# ISLP Chapter 9 Lab: Support Vector Machines

**Text:** *An Introduction to Statistical Learning with Applications in Python*, 9.6

A **maximal margin classifier** separates two classes with the hyperplane that leaves the
widest gap but it only exists when the classes are perfectly separable, and it is dangerously
sensitive to single observations. The **support vector classifier** relaxes this by allowing
margin violations, controlled by a cost parameter. The **support vector machine** extends it
further by replacing inner products with a kernel, giving non-linear boundaries.

## 9.6.1 Support Vector Classifier

### The role of `C`

The support vector classifier solves

$$\max_{\beta,\,\epsilon}\; M \quad \text{subject to} \quad
y_i(\beta_0 + \beta_1 x_{i1} + \dots + \beta_p x_{ip}) \ge M(1 - \epsilon_i),\;
\epsilon_i \ge 0,\; \sum_i \epsilon_i \le \text{budget},$$

where the $\epsilon_i$ are slack variables permitting margin violations.

**Be careful with the direction of `C` in scikit-learn.** In the textbook's formulation a larger
budget means a wider, more tolerant margin. In `sklearn`, `C` is the *penalty* on violations, so
the relationship inverts:

- **small `C`** → violations are cheap → **wide margin**, many support vectors, higher bias,
  lower variance
- **large `C`** → violations are expensive → **narrow margin**, few support vectors, lower bias,
  higher variance

`C` is the bias-variance dial, and the right value is an empirical question for
cross-validation.

### Simulated data

Two classes of 25 points each, with class `+1` shifted by 1 unit. The shift is small enough that
the classes overlap deliberately, since the interesting behaviour of `C` only appears when
perfect separation is impossible.

```python
rng = np.random.default_rng(1)
X = rng.standard_normal((50, 2))
y = np.array([-1]*25+[1]*25)
X[y==1] += 1
fig, ax = subplots(figsize=(8,8))
ax.scatter(X[:,0],
           X[:,1],
           c=y,
           cmap=cm.coolwarm);
```

The plot confirms the classes are **not** linearly separable there is no straight line with all
blue on one side and all red on the other. A maximal margin classifier would have no solution
here; the support vector classifier does, because it can buy violations.

```python
svm_linear = SVC(C=10, kernel='linear')
svm_linear.fit(X, y)
```

In the plot below, support vectors are marked `+` and all other observations are circles. Only
the support vectors matter: move any circle (without crossing the margin) and the boundary does
not shift at all. This is what makes SVMs robust to observations far from the boundary 
and the opposite of, say, LDA, where every point contributes to the class means.

```python
fig, ax = subplots(figsize=(8,8))
plot_svm(X,
         y,
         svm_linear,
         ax=ax)
```

Dropping to `C=0.1` makes violations cheap, so the margin widens and far more observations
become support vectors. Compare the two figures: same data, same kernel, very different
boundaries — and the wider-margin fit rests on more points, so it is the more stable of the two.

```python
svm_linear_small = SVC(C=0.1, kernel='linear')
svm_linear_small.fit(X, y)
fig, ax = subplots(figsize=(8,8))
plot_svm(X,
         y,
         svm_linear_small,
         ax=ax)
```

For a linear kernel the boundary coefficients are directly available. There is no such attribute
for non-linear kernels, where the boundary lives in the (implicit, possibly infinite-dimensional)
feature space and cannot be written down in the original coordinates one of the concrete costs
of kernelization.

```python
svm_linear.coef_
```

### Tuning `C` by cross-validation

`SVC` is an ordinary scikit-learn estimator, so `GridSearchCV` applies without modification.
`C=1` wins with a CV accuracy of 0.74, although several larger values tie exactly. When results
tie, prefer the **smallest** `C` that is the wider margin and the simpler model, and the tie
means nothing is being given up for it.

The two smallest values score 0.46, worse than chance. There the penalty is so weak that the
classifier essentially ignores the data.

```python
kfold = skm.KFold(5,
                  random_state=0,
                  shuffle=True)
grid = skm.GridSearchCV(svm_linear,
                        {'C':[0.001,0.01,0.1,1,5,10,100]},
                        refit=True,
                        cv=kfold,
                        scoring='accuracy')
grid.fit(X, y)
grid.best_params_
```

```python
grid.cv_results_[('mean_test_score')]
```

### Evaluating on new data

A fresh test set of 20 observations, generated the same way. Note this draws from the same `rng`
stream, so it is genuinely new data rather than a reshuffle of the training set.

```python
X_test = rng.standard_normal((20, 2))
y_test = np.array([-1]*10+[1]*10)
X_test[y_test==1] += 1
```

The CV-selected model gets 70% of test observations right. With 20 test points, that is
14 correct and a single flipped observation moves the figure by 5 percentage points, so
treat small differences here as noise.

```python
best_ = grid.best_estimator_
y_test_hat = best_.predict(X_test)
confusion_table(y_test_hat, y_test)
```

Using `C=0.001` instead gives 60%. The confusion table shows *how* it fails: almost everything
is pushed into class `1`. The margin is so wide that the classifier has nearly degenerated into
predicting one class which is why raw accuracy alone can hide a broken model, and the table
is worth printing every time.

```python
svm_ = SVC(C=0.001,
           kernel='linear').fit(X, y)
y_test_hat = svm_.predict(X_test)
confusion_table(y_test_hat, y_test)
```

### The separable case

Now push the classes apart until they are just barely separable. Note `X[y==1] += 1.9` **mutates
`X` in place** everything below uses the shifted data, and re-running this cell shifts it again.
Re-run from the top rather than piecemeal.

```python
X[y==1] += 1.9;
fig, ax = subplots(figsize=(8,8))
ax.scatter(X[:,0], X[:,1], c=y, cmap=cm.coolwarm);
```

With `C=1e5`, violations are effectively forbidden, so this approximates the maximal margin
classifier. Zero training errors, as expected.

```python
svm_ = SVC(C=1e5, kernel='linear').fit(X, y)
y_hat = svm_.predict(X)
confusion_table(y_hat, y)
```

Only **three** support vectors, all sitting exactly on the margin and defining it. That should
prompt scepticism rather than satisfaction: the entire classifier depends on three observations,
and moving any one of them swings the boundary. Perfect training accuracy achieved from three
points is a warning sign, not an achievement.

```python
fig, ax = subplots(figsize=(8,8))
plot_svm(X,
         y,
         svm_,
         ax=ax)
```

`C=0.1` still makes zero training errors, but now uses twelve support vectors and a much wider
margin. Same perfect training performance, far more stable boundary — and a larger test set
would be expected to favour this model.

This is the chapter's central lesson in miniature: when two models fit the training data equally
well, prefer the one resting on more evidence. Training accuracy cannot distinguish them; the
margin can.

```python
svm_ = SVC(C=0.1, kernel='linear').fit(X, y)
y_hat = svm_.predict(X)
confusion_table(y_hat, y)
```

```python
fig, ax = subplots(figsize=(8,8))
plot_svm(X,
         y,
         svm_,
         ax=ax)
```

## 9.6.2 Support Vector Machine

The support vector classifier depends on the data only through inner products
$\langle x_i, x_{i'}\rangle$. Replacing them with a **kernel** $K(x_i, x_{i'})$ gives a boundary
that is linear in an enlarged feature space and non-linear in the original one — without ever
computing the enlarged features. That is the kernel trick.

- Polynomial: $K(x_i,x_{i'}) = (1 + \sum_j x_{ij}x_{i'j})^d$ — `kernel='poly'`, `degree=d`
- Radial (RBF): $K(x_i,x_{i'}) = \exp(-\gamma \sum_j (x_{ij}-x_{i'j})^2)$ — `kernel='rbf'`, `gamma`

$\gamma$ controls how local the radial kernel is. Large $\gamma$ means distant points have
almost no influence, giving a wiggly boundary that can isolate individual observations; small
$\gamma$ gives a smoother, more global fit. With an RBF kernel there are **two** parameters to
tune, and they interact — `C` and `gamma` must be searched jointly, not one at a time.

### Data with a non-linear boundary

200 observations arranged so class 1 surrounds class 2 — no straight line can separate them.
Note the classes are unbalanced, 150 against 50.

```python
X = rng.standard_normal((200, 2))
X[:100] += 2
X[100:150] -= 2
y = np.array([1]*150+[2]*50)
```

```python
fig, ax = subplots(figsize=(8,8))
ax.scatter(X[:,0],
           X[:,1],
           c=y,
           cmap=cm.coolwarm)
```

Split in half, then fit a radial kernel with `gamma=1, C=1`.

One practical note the lab sidesteps because the data are simulated: SVMs are **not**
scale-invariant. Both the RBF kernel and the margin depend on Euclidean distances, so a
predictor measured in thousands will dominate one measured in units. On real data, standardize
inside a `Pipeline` so the scaler is fitted on training folds only.

```python
(X_train,
 X_test,
 y_train,
 y_test) = skm.train_test_split(X,
                                y,
                                test_size=0.5,
                                random_state=0)
svm_rbf = SVC(kernel="rbf", gamma=1, C=1)
svm_rbf.fit(X_train, y_train)
```

```python
fig, ax = subplots(figsize=(8,8))
plot_svm(X_train,
         y_train,
         svm_rbf,
         ax=ax)
```

The boundary is clearly non-linear, and a fair number of training points fall on the wrong side.
Raising `C` to `1e5` removes most of those errors — but look at the shape of the boundary it
produces: irregular, with narrow tendrils reaching out to capture individual observations. That
is overfitting made visible. Fewer training errors, worse expected test performance.

```python
svm_rbf = SVC(kernel="rbf", gamma=1, C=1e5)
svm_rbf.fit(X_train, y_train)
fig, ax = subplots(figsize=(8,8))
plot_svm(X_train,
         y_train,
         svm_rbf,
         ax=ax)
```

### Tuning `C` and `gamma` jointly

A 5×5 grid, scored by 5-fold CV on the training half only.

A caution on reading this cell: the printed result is `{'C': 100, 'gamma': 1}`, while the book's
surrounding text claims `C=1, gamma=0.5`. The text and its own output disagree. The text also
notes that several combinations tie, which is the likely explanation — with 100 training points
the CV accuracy surface is flat across many settings and the winner is effectively arbitrary.
Report whatever your own run produces, and check `grid.cv_results_['mean_test_score']` to see
how many settings tie before drawing any conclusion about the "best" parameters.

```python
kfold = skm.KFold(5,
                  random_state=0,
                  shuffle=True)
grid = skm.GridSearchCV(svm_rbf,
                        {'C':[0.1,1,10,100,1000],
                         'gamma':[0.5,1,2,3,4]},
                        refit=True,
                        cv=kfold,
                        scoring='accuracy');
grid.fit(X_train, y_train)
grid.best_params_
```

Worth inspecting directly — how flat is the surface?

```python
import pandas as pd
pd.DataFrame(grid.cv_results_)[['param_C', 'param_gamma', 'mean_test_score']] \
    .sort_values('mean_test_score', ascending=False).head(10)
```

The tuned model misclassifies about 12% of test observations. The boundary is smooth and
plausible — neither the underfit of the first RBF attempt nor the tendrils of the `C=1e5` fit.

```python
best_svm = grid.best_estimator_
fig, ax = subplots(figsize=(8,8))
plot_svm(X_train,
         y_train,
         best_svm,
         ax=ax)
y_hat_test = best_svm.predict(X_test)
confusion_table(y_hat_test, y_test)
```

## 9.6.3 ROC Curves

An SVM returns hard class labels, but underneath there is a continuous **fitted value** —
$\hat\beta_0 + \hat\beta_1 X_1 + \dots + \hat\beta_p X_p$ for the linear classifier, and the
kernel expansion (9.23) in general. Its *sign* gives the predicted class; its *magnitude* says
how far from the boundary the observation lies. `decision_function()` exposes it.

Shifting the threshold away from zero trades sensitivity against specificity. Sweeping it across
all values traces the **ROC curve**, and the area under it (AUC) summarizes performance across
every threshold at once — useful precisely because it does not depend on the somewhat arbitrary
choice of cutoff, and because it is insensitive to class imbalance in a way raw accuracy is not.

```python
fig, ax = subplots(figsize=(8,8))
roc_curve(best_svm,
          X_train,
          y_train,
          name='Training',
          #color='r',
          ax=ax);
```

Raising $\gamma$ to 50 makes the kernel extremely local. On the **training** data the ROC curve
improves and AUC approaches 1 — which sounds like progress and is not. Note the raw string
prefix `r'...'` on the label: without it, `\gamma` would be read as an escape sequence.

```python
svm_flex = SVC(kernel="rbf",
               gamma=50,
               C=1)
svm_flex.fit(X_train, y_train)
fig, ax = subplots(figsize=(8,8))
roc_curve(svm_flex,
          X_train,
          y_train,
          name=r'Training $\gamma=50$',
          #color='r',
          ax=ax);
```

Adding the **test** curve for the same $\gamma = 50$ model to the same axes tells the real story:
a large gap between the red training curve and the blue test curve. The flexible model has
memorized the training set. This is the cleanest diagnostic in the lab — an ROC curve computed
on training data is close to meaningless on its own, and only the train/test pair is informative.

```python
roc_curve(svm_flex,
          X_test,
          y_test,
          name=r'Test $\gamma=50$',
          #color='b',
          ax=ax)
fig;
```

The CV-tuned model, plotted the same way. The two curves sit much closer together — training
performance is a fair guide to test performance here, which is exactly what a well-tuned model
should look like. Compare the gap with the previous figure.

```python
fig, ax = subplots(figsize=(8,8))
for (X_, y_, c, name) in zip(
        (X_train, X_test),
        (y_train, y_test),
        ('r', 'b'),
        ('CV tuned on training',
         'CV tuned on test')):
    roc_curve(best_svm,
              X_,
              y_,
              name=name,
              ax=ax,
              #color=c
              )
```

## 9.6.4 SVM with Multiple Classes

The margin idea is defined for two classes, so multi-class problems are decomposed:

- **One-versus-one** (`'ovo'`): fit $\binom{K}{2}$ classifiers, one per pair, and assign the
  class winning the most pairwise contests. More models, each trained on less data.
- **One-versus-rest** (`'ovr'`): fit $K$ classifiers, each separating one class from all others,
  and assign the class with the largest fitted value. Fewer models, but each faces an unbalanced
  problem.

A third class is appended to the existing 200 observations.

```python
rng = np.random.default_rng(123)
X = np.vstack([X, rng.standard_normal((50, 2))])
y = np.hstack([y, [0]*50])
X[y==0,1] += 2
fig, ax = subplots(figsize=(8,8))
ax.scatter(X[:,0], X[:,1], c=y, cmap=cm.coolwarm);
```

`scatter_cmap=cm.tab10` gives distinguishable colors for three classes — `coolwarm` is a
two-ended gradient and makes the middle class hard to see.

```python
svm_rbf_3 = SVC(kernel="rbf",
                C=10,
                gamma=1,
                decision_function_shape='ovo');
svm_rbf_3.fit(X, y)
fig, ax = subplots(figsize=(8,8))
plot_svm(X,
         y,
         svm_rbf_3,
         scatter_cmap=cm.tab10,
         ax=ax)
```

For a quantitative response, `sklearn.svm.SVR` performs support vector regression — the same
margin idea, with the loss ignoring residuals smaller than a tolerance instead of classifying.

## 9.6.5 Application to Gene Expression Data

The `Khan` data holds expression measurements for four types of small round blue cell tumor.

```python
Khan = load_data('Khan')
Khan['xtrain'].shape, Khan['xtest'].shape
```

2,308 genes measured on 63 training and 20 test samples: $p \gg n$ by a factor of about 37.

This is precisely where SVMs earn their place, and it also settles the kernel choice. With far
more features than observations the classes are almost certainly separable by a hyperplane
already, so the extra flexibility of a polynomial or radial kernel buys nothing and only adds
variance. **Use a linear kernel.** More generally: when $p \gg n$, reach for the least flexible
model that can fit, not the most powerful one available.

```python
khan_linear = SVC(kernel='linear', C=10)
khan_linear.fit(Khan['xtrain'], Khan['ytrain'])
confusion_table(khan_linear.predict(Khan['xtrain']),
                Khan['ytrain'])
```

Zero training errors — and this is not evidence of a good model. With $p \gg n$ a separating
hyperplane essentially always exists, so perfect training classification is nearly guaranteed
regardless of whether the genes carry real signal. A permutation check (shuffling the labels and
refitting) would very likely also give zero training errors.

```python
confusion_table(khan_linear.predict(Khan['xtest']),
                Khan['ytest'])
```

Two test errors out of 20 — that number, unlike the training result, means something. With only
20 test samples the confidence interval around 90% accuracy is wide, so the honest statement is
that the classifier performs well on a small test set, not that its accuracy is 90%.

## Summary

**On the methods**

- The support vector classifier maximizes the margin while allowing violations. Only support
  vectors influence the boundary; every other observation could be moved without effect.
- `C` in `sklearn` penalizes violations, **inverting** the textbook's budget parameter: small `C`
  gives a wide margin and a stable, higher-bias fit; large `C` gives a narrow margin and risks
  overfitting.
- Kernels give non-linear boundaries by replacing inner products, never computing the enlarged
  features. The price is losing `coef_` there is no boundary to write down in the original
  coordinates.
- The RBF kernel needs `C` and `gamma` tuned **jointly**; they interact and a one-at-a-time
  search can miss the good region.
- ROC and AUC summarize performance across all thresholds, which makes them more informative
  than a single accuracy figure, especially with unbalanced classes.
- Multi-class is handled by decomposition into binary problems (`'ovo'` or `'ovr'`).

**What the simulations demonstrated**

- On separable data, `C=1e5` used three support vectors and `C=0.1` used twelve. Both made zero
  training errors; the second is the better model. Training accuracy cannot tell them apart.
- A `gamma=50` RBF looked near-perfect on training ROC and fell apart on test ROC the clearest
  overfitting signal in the lab.
- Perfect training classification on `Khan` follows from $p \gg n$ alone and is not evidence of
  a good model. The two test errors are the informative result.

**Practices to carry forward**

- Standardize features inside a `Pipeline` before fitting an SVM on real data; the method is
  distance-based and not scale-invariant.
- Print the confusion table alongside accuracy the `C=0.001` fit reached 60% while having
  nearly collapsed into predicting one class.
- When CV scores tie across settings, prefer the simpler model (smaller `C`, smaller `gamma`),
  and inspect `cv_results_` before claiming any single combination is best.
- Plot training and test ROC curves on the same axes. The gap between them, not the height of
  either, is the diagnostic.
