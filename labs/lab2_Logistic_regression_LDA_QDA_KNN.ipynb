## 4.7.2 Logistic Regression

**WHY `sm.GLM` and not sklearn.** Logistic regression is a generalized linear model:
a distribution for the response (here Bernoulli, passed as
`family=sm.families.Binomial()`) plus a **link function** transforming the mean into
something linear in the predictors. For logistic regression that link is the logit,

$$\eta(\mu)=\log\!\left(\frac{\mu}{1-\mu}\right)=\beta_0+\beta_1X_1+\dots+\beta_pX_p$$

`statsmodels` fits it by unpenalised maximum likelihood and reports standard errors,
z-statistics and p-values. `sklearn` reports none of those — it applies an L2 penalty by
default and targets prediction, not inference. Use the right tool for the question you are
asking. (`sm.Logit()` fits the same model directly; `sm.GLM` generalises to Poisson,
Gamma and the rest, which is why the lab uses it.)

```python
allvars = Smarket.columns.drop(['Today', 'Direction', 'Year'])
design = MS(allvars)
X = design.fit_transform(Smarket)
y = Smarket.Direction == 'Up'

glm = sm.GLM(y, X, family=sm.families.Binomial())
results = glm.fit()
summarize(results)
```

The smallest p-value belongs to `Lag1`, and its negative coefficient says a positive return
yesterday makes an increase today slightly *less* likely. But at 0.145 that p-value is not
close to significant, so there is no real evidence of any association.

**Note on the response.** `y` is a Boolean (`Direction == 'Up'`), so `True`/`Up` is the
event being modelled. If you flip the coding, every coefficient flips sign and the fitted
probabilities become `1 - p`. Always state which class is "1".

```python
results.params
```

```python
results.pvalues.round(4)
```

`results.predict()` returns fitted values **on the probability scale** (not log odds).
With no argument it predicts on the training data.

```python
probs = results.predict()
probs[:10].round(4)
```

To turn probabilities into class labels we need a **threshold**. 0.5 is the conventional
default, not a property of the model — §4.7.6 shows what happens when 0.5 is the wrong
choice.

```python
labels = np.array(['Down'] * 1250)
labels[probs > 0.5] = 'Up'

confusion_table(labels, Smarket.Direction)
```

`confusion_table` puts **predictions down the rows and truth across the columns** — the
transpose of `sklearn.metrics.confusion_matrix`. Diagonal entries are correct. So the model
got 507 `Up` days and 145 `Down` days right.

```python
print((507 + 145) / 1250, np.mean(labels == Smarket.Direction))
```

> **WHY this 52.2% is meaningless.** We trained and tested on the same 1,250 days, so
> 47.8% is the **training** error rate, which systematically underestimates test error.
> The fix is a genuine held-out set. Because this is time series data, we hold out by
> *time* rather than at random: train on 2001–2004, test on 2005. Randomly shuffling days
> would let the model peek at the future, which is exactly the mistake a trading strategy
> cannot afford.

```python
train = (Smarket.Year < 2005)
Smarket_train = Smarket.loc[train]
Smarket_test = Smarket.loc[~train]
Smarket_test.shape
```

`train` is a Boolean array: `True` before 2005, `False` in 2005. The `~` operator negates
it, so `Smarket.loc[~train]` selects the 252 days of 2005.

(The book prints this as `∼train`. That is a typeset tilde — type the ASCII `~`.)

```python
X_train, X_test = X.loc[train], X.loc[~train]
y_train, y_test = y.loc[train], y.loc[~train]

glm_train = sm.GLM(y_train, X_train, family=sm.families.Binomial())
results = glm_train.fit()
probs = results.predict(exog=X_test)

D = Smarket.Direction
L_train, L_test = D.loc[train], D.loc[~train]

labels = np.array(['Down'] * 252)
labels[probs > 0.5] = 'Up'
confusion_table(labels, L_test)
```

```python
print("accuracy:", np.mean(labels == L_test).round(4),
      "| error rate:", np.mean(labels != L_test).round(4))
```

A test error rate of 52% — **worse than a coin flip**. Which is the honest answer: previous
days' returns do not predict tomorrow's market.

> **WHY dropping predictors can help.** Predictors with no true relationship to the
> response still add variance to the fit without reducing bias, so they push test error up.
> Refitting with only `Lag1` and `Lag2` (the two with the largest, if unimpressive, signal)
> should therefore do no worse.

```python
model = MS(['Lag1', 'Lag2']).fit(Smarket)
X = model.transform(Smarket)
X_train, X_test = X.loc[train], X.loc[~train]

glm_train = sm.GLM(y_train, X_train, family=sm.families.Binomial())
results = glm_train.fit()
probs = results.predict(exog=X_test)

labels = np.array(['Down'] * 252)
labels[probs > 0.5] = 'Up'
confusion_table(labels, L_test)
```

```python
print("overall accuracy      :", round((35 + 106) / 252, 4))
print("accuracy when 'Up'    :", round(106 / (106 + 76), 4))
print("always-predict-Up rate:", round((L_test == 'Up').mean(), 4))
```

Read this carefully — it is the single most instructive result in the lab.

56% overall looks like progress, **but always predicting `Up` also scores 56%**, because
the market rose on 56% of days in 2005. In overall accuracy the model adds nothing.

What *is* interesting: on the days it predicts an increase it is right 58% of the time.
That suggests a strategy — trade on predicted-up days, sit out the rest. Whether a 2-point
edge on 182 days is real or noise is exactly the kind of question a hypothesis test exists
to answer, and 182 observations will not settle it.

**The general lesson:** always compare accuracy against the majority-class rate, and look
inside the confusion table rather than at the single accuracy number.

```python
newdata = pd.DataFrame({'Lag1': [1.2, 1.5], 'Lag2': [1.1, -0.8]})
newX = model.transform(newdata)
results.predict(newX)
```

Both below 0.5, so both days would be predicted `Down`.

Note we passed `newdata` through `model.transform()`. The `ModelSpec` remembers the design
(which columns, the intercept, any contrast coding), so new data is built exactly the way
the training matrix was. Constructing it by hand is how column-order bugs happen.

## 4.7.3 Linear Discriminant Analysis

**WHY a second method.** Logistic regression models $\Pr(Y=k\mid X)$ directly. LDA goes the
other way: it models the distribution of $X$ *within* each class as multivariate normal with
a **common covariance matrix**, then flips that around with Bayes' theorem:

$$\Pr(Y=k\mid X=x)=\frac{\pi_k f_k(x)}{\sum_{l}\pi_l f_l(x)}$$

The shared covariance is what makes the decision boundary linear. LDA earns its keep when
the classes are well separated (logistic regression's estimates become unstable there), when
$n$ is small and normality is plausible, and when there are more than two classes.

```python
lda = LDA(store_covariance=True)

# LDA adds its own intercept, so drop the one ModelSpec created.
# We also pass the labels ('Up'/'Down') directly rather than the Boolean y.
X_train, X_test = [M.drop(columns=['intercept']) for M in [X_train, X_test]]
lda.fit(X_train, L_train)
```

```python
print("classes_:", lda.classes_)
print("priors_ :", lda.priors_.round(3))       # pi-hat_k
print("means_  :\n", np.round(lda.means_, 3))   # mu-hat_k, one row per class
```

The class means say the previous two days' returns tend to be *negative* before an up day
and *positive* before a down day — a weak mean-reversion signal, consistent with the
negative `Lag1` coefficient from the logistic fit.

The trailing underscore is an sklearn convention: it marks a quantity **estimated during
`fit()`**, as opposed to a hyperparameter you set yourself.

```python
lda.scalings_
```

These are the multipliers in the linear discriminant. When
$-0.64\times\texttt{Lag1}-0.51\times\texttt{Lag2}$ is large, LDA predicts an increase.

```python
lda_pred = lda.predict(X_test)
confusion_table(lda_pred, L_test)
```

**Identical to logistic regression.** Same 35/35/76/106 table. This is the theoretical
point from §4.5 showing up empirically: both methods produce linear boundaries and differ
only in how they estimate them (maximum likelihood versus estimated class means and a
pooled covariance). When the normality assumption is roughly satisfied, they agree.

```python
lda_prob = lda.predict_proba(X_test)

# Applying a 50% cut-off to the posteriors recreates .predict() exactly:
print(np.all(np.where(lda_prob[:, 1] >= 0.5, 'Up', 'Down') == lda_pred))

# For more than two classes, .predict() takes the argmax of the posteriors:
print(np.all([lda.classes_[i] for i in np.argmax(lda_prob, 1)] == lda_pred))
```

```python
# Column 0 is 'Down' (check lda.classes_). How many days are we >90% sure will fall?
print("days with P(Down) > 0.90:", np.sum(lda_prob[:, 0] > 0.9))
print("largest P(Down) in 2005 :", lda_prob[:, 0].max().round(4))
```

Not one day in 2005 clears 90% — the most confident prediction of a decline was 52%. The
model is barely distinguishable from a coin, which is a far more useful thing to know than
its accuracy.

## 4.7.4 Quadratic Discriminant Analysis

**WHY QDA.** Drop LDA's assumption of a *shared* covariance matrix and give each class its
own $\Sigma_k$. The $x^\top\Sigma^{-1}x$ terms no longer cancel, so the boundary becomes
quadratic. That buys flexibility at the cost of many more parameters
($K p(p+1)/2$ instead of $p(p+1)/2$), so QDA needs more data — the bias-variance trade-off
in its purest classroom form.

```python
qda = QDA(store_covariance=True)
qda.fit(X_train, L_train)

print("means_ :\n", np.round(qda.means_, 4))
print("priors_:", qda.priors_.round(4))
```

```python
# One covariance matrix per class -- here is the first ('Down'):
qda.covariance_[0].round(4)
```

There are no "linear discriminant coefficients" to print, because the discriminant is
quadratic rather than linear.

```python
qda_pred = qda.predict(X_test)
print(confusion_table(qda_pred, L_test))
print("\naccuracy:", np.mean(qda_pred == L_test).round(4))
```

**60% — the best result in the lab**, on data no one expects to be predictable, from a
model fit without ever seeing 2005. That hints the true boundary really is curved.

Treat it with suspicion anyway: it is one 252-day test set, and the gap over the
always-predict-`Up` rate of 56% is about four percentage points. The book's own advice is to
check on a larger test set before betting real money.

## 4.7.5 Naive Bayes

**WHY "naive".** Naive Bayes also models $f_k(x)$, but assumes the predictors are
**independent within each class**, so the joint density factorises into a product of
one-dimensional densities. That is usually false, yet it works well in practice: it
replaces a $p\times p$ covariance matrix with $p$ variances, which slashes variance at the
cost of some bias. `GaussianNB` uses a normal density for each feature; kernel density
estimates are also possible.

```python
NB = GaussianNB()
NB.fit(X_train, L_train)

print("classes_    :", NB.classes_)
print("class_prior_:", NB.class_prior_.round(4))
print("theta_ (means):\n", np.round(NB.theta_, 4))
print("var_          :\n", np.round(NB.var_, 4))
```

Rows are classes, columns are features. Verify the estimates by hand — always worth doing
once so the attribute names stop being magic:

```python
print(X_train[L_train == 'Down'].mean())
print()
print(X_train[L_train == 'Down'].var(ddof=0))   # ddof=0 -> the 1/n convention
```

> **Watch the variance convention.** `GaussianNB` divides by $n$; pandas' `.var()` divides
> by $n-1$ by default. They agree only if you pass `ddof=0`. The same 1/n-vs-1/(n-1) issue
> reappears with `StandardScaler` in §4.7.6.

```python
nb_labels = NB.predict(X_test)
print(confusion_table(nb_labels, L_test))
print("\naccuracy:", np.mean(nb_labels == L_test).round(4))
```

```python
NB.predict_proba(X_test)[:5].round(4)
```

About 59.5% — a shade below QDA, well above LDA. With only two predictors the independence
assumption costs little; with many correlated predictors it would cost more.

## 4.7.6 K-Nearest Neighbours

**WHY KNN is different.** Every method so far was *parametric*: assume a form, estimate its
parameters. KNN assumes nothing — it finds the $K$ closest training points and takes a
majority vote. $K$ controls flexibility directly: $K=1$ is maximally flexible (zero training
error, high variance), and larger $K$ smooths the boundary.

```python
knn1 = KNeighborsClassifier(n_neighbors=1)
knn1.fit(X_train, L_train)
knn1_pred = knn1.predict(X_test)
print(confusion_table(knn1_pred, L_test))
print("\nK=1 accuracy:", np.mean(knn1_pred == L_test).round(4))
```

```python
knn3_pred = KNeighborsClassifier(n_neighbors=3).fit(X_train, L_train).predict(X_test)
print("K=3 accuracy:", np.mean(knn3_pred == L_test).round(4))
```

$K=1$ lands on exactly 50% — a coin flip — because it fits the training noise perfectly and
generalises none of it. $K=3$ improves slightly; going further does not help.

### Scoreboard for `Smarket`

```python
scores = {
    'Logistic (Lag1+Lag2)': np.mean(labels == L_test),
    'LDA': np.mean(lda_pred == L_test),
    'QDA': np.mean(qda_pred == L_test),
    'Naive Bayes': np.mean(nb_labels == L_test),
    'KNN (K=1)': np.mean(knn1_pred == L_test),
    'KNN (K=3)': np.mean(knn3_pred == L_test),
    'Always predict "Up"': (L_test == 'Up').mean(),
}
pd.DataFrame({'Test accuracy (2005)': pd.Series(scores).round(4)}) \
  .sort_values('Test accuracy (2005)', ascending=False)
```

Logistic regression and LDA tie exactly (linear boundaries). QDA and Naive Bayes clear the
always-`Up` baseline; the linear methods do not. KNN is worst — with only two noisy
predictors, local averaging has nothing to average.

### KNN on the `Caravan` data

`Smarket` makes KNN look useless. `Caravan` shows what it can do: 85 demographic predictors
for 5,822 people, response `Purchase` (did they buy caravan insurance). Only about 6% did —
a **badly imbalanced** problem, which is the real lesson of this section.

```python
Caravan = load_data('Caravan')
Purchase = Caravan.Purchase
print(Purchase.value_counts())
print("\npurchase rate:", round(348 / 5822, 4))
```

> **WHY scaling is mandatory for KNN — and not for logistic regression.** KNN's only input
> is distance. A variable measured in thousands dominates one measured in single digits
> purely because of its units: a 1,000-unit salary gap swamps a 50-year age gap. Measure
> salary in yen and the classifier changes its mind. Standardising puts every variable on
> mean 0, SD 1 so they contribute comparably.
>
> Contrast with logistic regression, where rescaling a predictor just divides its
> coefficient by the same factor and leaves fitted probabilities untouched — scaling matters
> there only when you penalise.

```python
feature_df = Caravan.drop(columns=['Purchase'])

scaler = StandardScaler(with_mean=True, with_std=True, copy=True)
scaler.fit(feature_df)
X_std = scaler.transform(feature_df)

feature_std = pd.DataFrame(X_std, columns=feature_df.columns)
feature_std.std().head()
```

The SDs read 1.000086 rather than exactly 1 — the 1/n versus 1/(n−1) convention again
(`StandardScaler` uses 1/n, `.std()` uses 1/(n−1)). Harmless, as long as every column is
treated the same way.

```python
(X_train, X_test, y_train, y_test) = train_test_split(feature_std, Purchase,
                                                      test_size=1000, random_state=0)

knn1 = KNeighborsClassifier(n_neighbors=1)
knn1_pred = knn1.fit(X_train, y_train).predict(X_test)

print("KNN error rate :", np.mean(y_test != knn1_pred).round(4))
print("null rate      :", np.mean(y_test != "No").round(4))
```

> **The null rate.** An 11% error rate sounds respectable until you notice that predicting
> `No` for everyone gives about 6%. **KNN is worse than doing nothing** by that measure.
>
> But overall error is the wrong measure here. If a salesperson must visit each prospect,
> what matters is the hit rate *among those we approach* — precision on the `Yes`
> predictions, not accuracy overall. Choose the metric from the cost structure, not out of
> habit.

```python
confusion_table(knn1_pred, y_test)
```

```python
print("hit rate among predicted buyers:", round(9 / (53 + 9), 4))
print("hit rate from random guessing  :", round((y_test == 'Yes').mean(), 4))
```

Among the 62 people KNN flags, 9 actually buy — **14.5%, more than double the 6% base
rate**. By the metric that matters, KNN is genuinely useful on data where its overall
accuracy looked like a failure.

### Tuning K

$K$ is a **tuning parameter** (hyperparameter): unknown in advance, chosen by looking at
performance. Here we sweep it against hit rate among predicted buyers.

```python
for K in range(1, 6):
    knn = KNeighborsClassifier(n_neighbors=K)
    knn_pred = knn.fit(X_train, y_train).predict(X_test)
    C = confusion_table(knn_pred, y_test)
    pred = C.loc['Yes'].sum()
    did_buy = C.loc['Yes', 'Yes']
    rate = did_buy / pred if pred else float('nan')
    print(f"K={K}: # predicted to buy: {pred:>2}, # who did buy {did_buy}, accuracy {rate:.1%}")
```

The counts are tiny and jump around — at $K=4$ only a handful are flagged and none buy. Do
not read a trend into this; with 3–62 predictions per setting, these rates carry enormous
sampling error.

> **Method note.** Tuning $K$ by staring at the *test* set is exactly the mistake that makes
> reported scores optimistic. It is done here to keep the lab simple; Chapter 5 replaces it
> with cross-validation on the training set.

> **ERRATUM (In [60]).** The book's f-string says "predicted to rent" — leftover from the
> Bikeshare data. It should read "buy". And its $K=4$ row prints 3 predictions where current
> sklearn gives 4 (tie-breaking among even $K$ changed); the 0.0% accuracy is unaffected.

### Comparison to logistic regression

> **ERRATUM (In [61]) — the important one.** The book prints
> `np.where(logit_pred[:,1] > 5, 'Yes', 'No')`. A probability can never exceed 5, so that
> line predicts `No` for everyone by construction, which is where the all-zero `Yes` row in
> the book's output comes from. The intended threshold is **0.5**, as the surrounding text
> says. With the correct 0.5 you get 2 predicted buyers (both wrong) rather than 0 — the
> narrative survives, the numbers do not.

```python
# C=1e10 makes the L2 penalty negligible, recovering ordinary unpenalised logistic
# regression -- sklearn's default C=1 is closer to ridge regression (Chapter 6).
logit = LogisticRegression(C=1e10, solver='liblinear')
logit.fit(X_train, y_train)
logit_pred = logit.predict_proba(X_test)

logit_labels = np.where(logit_pred[:, 1] > 0.5, 'Yes', 'No')
print("Threshold 0.50 (corrected from the book's '> 5'):")
print(confusion_table(logit_labels, y_test))
```

At a 0.5 cut-off the model is useless for the business question: almost nobody is flagged,
and nobody flagged actually buys. That is not a broken model — it is a **badly chosen
threshold**. With a 6% base rate, few people ever cross 50%.

Lower the bar to 0.25:

```python
logit_labels = np.where(logit_pred[:, 1] > 0.25, 'Yes', 'No')
print(confusion_table(logit_labels, y_test))
print("\nhit rate among predicted buyers:", round(9 / (20 + 9), 4))
```

**31% of the people we approach buy — roughly five times the 6% base rate**, and better
than KNN's 14.5%. Same model, same coefficients, same data; only the threshold moved.

This is the practical face of the point made back in §4.3 of the text: a conservative lender
might flag defaults at $p>0.1$ rather than 0.5. The model gives you probabilities; **the
threshold is a business decision, not a statistical one.**

## 4.7.7 Linear and Poisson Regression on the Bikeshare Data

**WHY this belongs in a classification lab.** It shows the GLM recipe generalising. Swap the
family and the link and the same machinery handles counts:

| Model | Distribution | Link $\eta(\mu)$ |
|---|---|---|
| Linear | Gaussian | $\mu$ |
| Logistic | Bernoulli | $\log(\mu/(1-\mu))$ |
| Poisson | Poisson | $\log(\mu)$ |

`bikers` counts hourly bike rentals in Washington DC. Counts are non-negative integers, so
least squares can predict impossible negative ridership — the same defect that ruled out
linear regression for binary responses in §4.2.

```python
Bike = load_data('Bikeshare')
Bike.shape, list(Bike.columns)
```

```python
X = MS(['mnth', 'hr', 'workingday', 'temp', 'weathersit']).fit_transform(Bike)
Y = Bike['bikers']
M_lm = sm.OLS(Y, X).fit()
summarize(M_lm).head(14)
```

With 24 levels of `hr` and 12 of `mnth` this table runs to 40 rows. Under the default
**dummy coding**, the first level of each factor (`hr[0]`, `mnth[Jan]`) is the baseline with
an implicit coefficient of 0, and every other level is measured relative to it. So
`mnth[Feb] = 6.85` means roughly 7 more riders in February than in January.

```python
hr_encode = contrast('hr', 'sum')
mnth_encode = contrast('mnth', 'sum')

X2 = MS([mnth_encode, hr_encode, 'workingday', 'temp', 'weathersit']).fit_transform(Bike)
M2_lm = sm.OLS(Y, X2).fit()
S2 = summarize(M2_lm)
S2.head(14)
```

**Sum-to-zero coding** instead leaves out the *last* level, whose coefficient is minus the
sum of all the others. The coefficients therefore sum to zero and read as deviations from
the yearly average: January's −46.09 means about 46 fewer riders than an average month.

**WHY it does not matter.** The coding changes what the coefficients *mean*, never what the
model *predicts*. Same column space, same fit:

```python
print("sum of squared differences:", np.sum((M_lm.fittedvalues - M2_lm.fittedvalues) ** 2))
print("np.allclose:", np.allclose(M_lm.fittedvalues, M2_lm.fittedvalues))
```

Zero to machine precision. A favourite exam question: *does the choice of dummy coding
change the fitted values?* No — only the interpretation of individual coefficients.

Now reconstruct the omitted December coefficient and plot the seasonal pattern.

```python
coef_month = S2[S2.index.str.contains('mnth')]['coef']
coef_month = pd.concat([coef_month, pd.Series([-coef_month.sum()], index=['mnth[Dec]'])])
coef_month.round(4)
```

```python
fig_month, ax_month = subplots(figsize=(8, 8))
x_month = np.arange(coef_month.shape[0])
ax_month.plot(x_month, coef_month, marker='o', ms=10)
ax_month.set_xticks(x_month)
ax_month.set_xticklabels([l[5] for l in coef_month.index], fontsize=20)
ax_month.set_xlabel('Month', fontsize=20)
ax_month.set_ylabel('Coefficient', fontsize=20);
```

```python
coef_hr = S2[S2.index.str.contains('hr')]['coef']
coef_hr = coef_hr.reindex(['hr[{0}]'.format(h) for h in range(23)])
coef_hr = pd.concat([coef_hr, pd.Series([-coef_hr.sum()], index=['hr[23]'])])

fig_hr, ax_hr = subplots(figsize=(8, 8))
x_hr = np.arange(coef_hr.shape[0])
ax_hr.plot(x_hr, coef_hr, marker='o', ms=10)
ax_hr.set_xticks(x_hr[::2])
ax_hr.set_xticklabels(range(24)[::2], fontsize=20)
ax_hr.set_xlabel('Hour', fontsize=20)
ax_hr.set_ylabel('Coefficient', fontsize=20);
```

Two clean commuter peaks around 8am and 5–6pm, and a summer/autumn seasonal hump.

### Poisson regression

Almost nothing changes — swap the family.

```python
M_pois = sm.GLM(Y, X2, family=sm.families.Poisson()).fit()
summarize(M_pois).head(8)
```

```python
S_pois = summarize(M_pois)

coef_month = S_pois[S_pois.index.str.contains('mnth')]['coef']
coef_month = pd.concat([coef_month, pd.Series([-coef_month.sum()], index=['mnth[Dec]'])])

coef_hr = S_pois[S_pois.index.str.contains('hr')]['coef']
coef_hr = coef_hr.reindex(['hr[{0}]'.format(h) for h in range(23)])
coef_hr = pd.concat([coef_hr, pd.Series([-coef_hr.sum()], index=['hr[23]'])])

fig_pois, (ax_month, ax_hr) = subplots(1, 2, figsize=(16, 8))

ax_month.plot(x_month, coef_month, marker='o', ms=10)
ax_month.set_xticks(x_month)
ax_month.set_xticklabels([l[5] for l in coef_month.index], fontsize=20)
ax_month.set_xlabel('Month', fontsize=20)
ax_month.set_ylabel('Coefficient', fontsize=20)

ax_hr.plot(x_hr, coef_hr, marker='o', ms=10)
ax_hr.set_xticks(x_hr[::2])                       # book omits this line -- see errata
ax_hr.set_xticklabels(range(24)[::2], fontsize=20)
ax_hr.set_xlabel('Hour', fontsize=20)
ax_hr.set_ylabel('Coefficient', fontsize=20);
```

Same shapes, different vertical scale: Poisson coefficients are on the **log** scale, so
$e^{\beta}$ is a *multiplicative* effect on expected ridership rather than an additive one.

```python
fig, ax = subplots(figsize=(8, 8))
ax.scatter(M2_lm.fittedvalues, M_pois.fittedvalues, s=20)
ax.set_xlabel('Linear Regression Fit', fontsize=20)
ax.set_ylabel('Poisson Regression Fit', fontsize=20)
ax.axline([0, 0], c='black', linewidth=3, linestyle='--', slope=1);
```

```python
print("negative predictions, linear model :", (M2_lm.fittedvalues < 0).sum())
print("negative predictions, Poisson model:", (M_pois.fittedvalues < 0).sum())
```

There it is: the linear model predicts **negative numbers of cyclists** for hundreds of
hours. The Poisson fit cannot, because the log link guarantees $\mu=e^{\eta}>0$. That is the
same argument that motivated logistic regression over linear regression for a binary
response in §4.2 — and it is why both belong to one family.

## Exam-oriented summary

**Choosing a method**

| Situation | Reach for |
|---|---|
| Need p-values, SEs, interpretable coefficients | Logistic regression (`statsmodels`) |
| Classes well separated, or $n$ small with roughly normal predictors | LDA |
| More than two classes | LDA or multinomial logistic |
| Boundary looks curved and $n$ is large | QDA |
| Very many predictors, little data | Naive Bayes |
| Boundary highly irregular, $n$ large, few predictors | KNN |
| Response is a count | Poisson regression |

**Points that recur in exams**

1. Training error flatters. 52.2% on `Smarket` became 48% on held-out data.
2. Always quote the **null rate** (majority-class / no-purchase rate) beside accuracy.
3. **The threshold is a choice.** Moving it from 0.5 to 0.25 took the `Caravan` hit rate
   from nothing to 31%. It does not change AUC.
4. **Logistic regression and LDA give near-identical predictions** — both linear boundaries,
   different estimation.
5. **LDA vs QDA is bias-variance.** Shared $\Sigma$ = fewer parameters, linear boundary;
   per-class $\Sigma_k$ = flexible, quadratic, data-hungry.
6. **Scaling matters for KNN, not for unpenalised logistic regression.** Distance versus
   coefficients.
7. **Dummy coding does not change fitted values**, only coefficient interpretation.
8. Tuning parameters ($K$, $C$) should be chosen by **cross-validation**, not on the test set.
