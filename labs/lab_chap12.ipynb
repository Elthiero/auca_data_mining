## 12.5.1 Principal Components Analysis

We perform PCA on the `USArrests` dataset.

`USArrests` is an R dataset: arrest statistics per 100,000 residents for each of the 50 US states, plus the percentage of the population living in urban areas. The rows are the 50 states in alphabetical order.

```python
USArrests = get_rdataset('USArrests').data
USArrests
```

Four variables: `Murder`, `Assault`, `UrbanPop`, and `Rape`.

```python
USArrests.columns
```

A quick look at the data. The column means are on completely different scales.

```python
USArrests.mean()
```

The variances differ even more dramatically. `Assault` has a variance of roughly 6,945,
against about 19 for `Murder`.

```python
USArrests.var()
```

**Why scaling matters here.** PCA searches for the directions that account for as much variance as possible. If we feed it the raw data, `Assault` whose variance is two to three orders of magnitude larger than the others will essentially *be* the first principal component, not because it is more informative but because it is measured on a bigger scale. `UrbanPop` is a percentage; the other three are counts per 100,000 people. These are not comparable numbers.

So whenever the variables are in different units or vary widely in scale, standardize them to standard deviation one before running PCA. Typically we center them at zero too.

`StandardScaler` follows the usual scikit-learn pattern: `fit()` computes the means and
standard deviations, `transform()` applies them, and `fit_transform()` does both in one
call.

```python
scaler = StandardScaler(with_std=True, with_mean=True)
USArrests_scaled = scaler.fit_transform(USArrests)
```

Now the PCA itself. By default `PCA()` centers the variables but does **not** scale them, which is exactly why we scaled the data ourselves in the previous step.

Calling `fit()` computes the loadings and populates the object's attributes.

```python
pcaUS = PCA()
pcaUS.fit(USArrests_scaled)
```

`mean_` holds the column means observed during `fit`. Since we already centered the data, these are all zero (up to floating-point noise, hence the signed zeros).

```python
pcaUS.mean_
```

`transform()` projects the observations onto the components. The result the *scores* has one row per state and one column per principal component.

```python
scores = pcaUS.transform(USArrests_scaled)
```

`components_` holds the loading vectors, one per row. Each row is a unit vector giving the
weight of each original variable in that component.

The first component loads roughly evenly on `Murder`, `Assault`, and `Rape` it is an overall crime-rate axis. The second loads heavily (and negatively) on `UrbanPop`, so it mostly separates urban from rural states.

Note that the sign of a principal component is arbitrary: flipping a whole loading vector and the matching score column describes exactly the same solution.

```python
pcaUS.components_
```

### Biplot (manual)

A **biplot** overlays two things on one figure: the observations plotted in the space of the first two score vectors, and an arrow per original variable pointing in the direction of its loadings. It lets you read off both which states are similar and what the axes mean.

scikit-learn has no built-in biplot, so we draw one by hand: scatter the first two score columns, then draw an arrow from the origin to each variable's `(PC1, PC2)` loading and label it.

```python
i, j = 0, 1  # which components
fig, ax = plt.subplots(1, 1, figsize=(8, 8))
ax.scatter(scores[:, 0], scores[:, 1])
ax.set_xlabel('PC%d' % (i+1))
ax.set_ylabel('PC%d' % (j+1))
for k in range(pcaUS.components_.shape[1]):
    ax.arrow(0, 0, pcaUS.components_[i, k], pcaUS.components_[j, k],
             head_width=0.05, head_length=0.05, color='r')
    ax.text(pcaUS.components_[i, k],
            pcaUS.components_[j, k],
            USArrests.columns[k])
```

Because the sign of each component is arbitrary, the figure above is a mirror image of Figure 12.1 in the book. Flipping the sign of the second score column *and* the second loading row reproduces the published orientation the clustering of the points is unchanged, only the picture is reflected. `s_` simply lengthens the arrows so the loadings are easier to read.

⚠️ This cell mutates `scores` and `pcaUS.components_` **in place**. If you re-run it, the signs flip back. Re-run the fit and transform cells above to get a clean state.

```python
scale_arrow = s_ = 2
scores[:, 1] *= -1
pcaUS.components_[1] *= -1  # flip the y-axis
fig, ax = plt.subplots(1, 1, figsize=(8, 8))
ax.scatter(scores[:, 0], scores[:, 1])
ax.set_xlabel('PC%d' % (i+1))
ax.set_ylabel('PC%d' % (j+1))
for k in range(pcaUS.components_.shape[1]):
    ax.arrow(0, 0,
             s_*pcaUS.components_[i, k],
             s_*pcaUS.components_[j, k],
             head_width=0.05, head_length=0.05, color='r')
    ax.text(s_*pcaUS.components_[i, k],
            s_*pcaUS.components_[j, k],
            USArrests.columns[k])
```

## 12.5.2 Matrix Completion

We recreate the USArrests analysis using SVD.

Matrix completion is the problem of filling in missing entries of a data matrix. The book shows (Section 12.2.2) that solving optimization problem (12.6) on a centered matrix is equivalent to computing its first *M* principal components, and the **singular value decomposition (SVD)** is the general algorithm for doing so.

`np.linalg.svd()` factors `X` into `U`, `D`, and `V`. With `full_matrices=False`, `U` has the same shape as `X` for a tall matrix.

```python
X = USArrests_scaled
U, D, V = np.linalg.svd(X, full_matrices=False)
U.shape, D.shape, V.shape
```

`V` is the loading matrix the same thing PCA returned, up to a sign flip on some rows.

```python
V
```

Compare with the PCA loadings: identical apart from the sign of the first row (and the second row, which we flipped by hand earlier).

```python
pcaUS.components_
```

`U` is a standardized version of the score matrix each column has sum of squares one. Multiplying each column by the corresponding singular value in `D` recovers the PCA scores.

```python
(U * D[None, :])[:3]
```

Same numbers as the PCA scores, up to the meaningless sign flip.

```python
scores[:3]
```

Now we deliberately create missing data: 20 entries knocked out of the 50 × 4 matrix.

We pick 20 distinct rows (`replace=False`) and, for each, one column at random
(`replace=True`, since columns can repeat across rows). Selecting distinct rows guarantees
every state keeps at least three observed values a row with everything missing could not
be imputed.

```python
n_omit = 20
np.random.seed(15)
r_idx = np.random.choice(np.arange(X.shape[0]), n_omit, replace=False)
c_idx = np.random.choice(np.arange(X.shape[1]), n_omit, replace=True)
Xna = X.copy()
Xna[r_idx, c_idx] = np.nan
```

This helper implements **Step 2(a)** of Algorithm 12.1: given a complete matrix, return its
best rank-`M` approximation via the SVD, keeping only the first `M` singular vectors and
values.

```python
def low_rank(X, M=1):
    U, D, V = np.linalg.svd(X)
    L = U[:, :M] * D[None, :M]
    return L.dot(V[:M])
```

**Step 1:** initialize the missing entries. `np.nanmean()` computes column means over the
observed values only, and we drop those means into the missing slots as a starting guess.
We copy `Xna` first so the original matrix with its NaNs is preserved.

```python
Xhat = Xna.copy()
Xbar = np.nanmean(Xhat, axis=0)
Xhat[r_idx, c_idx] = Xbar[c_idx]
```

Bookkeeping before the loop:

- `ismiss` is a boolean mask that is `True` where `Xna` is missing; `~ismiss` gives us the
  observed entries.
- `mss0` is the mean squared value of the observed entries the baseline.
- `mssold` tracks the previous iteration's mean squared error on the observed entries.

We iterate until the relative error `(mssold - mss) / mss0` falls below `thresh`. Dividing
by `mss0` is what makes the number of iterations independent of any rescaling of the raw
data.

```python
thresh = 1e-7
rel_err = 1
count = 0
ismiss = np.isnan(Xna)
mssold = np.mean(Xhat[~ismiss]**2)
mss0 = np.mean(Xna[~ismiss]**2)
```

**Step 2**, repeated until convergence:

- **(a)** approximate the current filled-in matrix with a rank-1 reconstruction (`Xapp`);
- **(b)** overwrite *only the missing entries* of `Xhat` with the corresponding values from
  `Xapp` the observed entries are never touched;
- **(c)** recompute the error on the observed entries and the relative improvement.

The algorithm converges after about eight iterations, settling at a mean squared error of
0.381 on the non-missing elements.

```python
while rel_err > thresh:
    count += 1
    # Step 2(a)
    Xapp = low_rank(Xhat, M=1)
    # Step 2(b)
    Xhat[ismiss] = Xapp[ismiss]
    # Step 2(c)
    mss = np.mean(((Xna - Xapp)[~ismiss])**2)
    rel_err = (mssold - mss) / mss0
    mssold = mss
    print("Iteration: {0}, MSS:{1:.3f}, Rel.Err {2:.2e}"
          .format(count, mss, rel_err))
```

The real test: how close are the 20 imputed values to the truth we hid? A correlation of
about 0.71 is respectable for a rank-1 approximation of a four-column matrix.

This implementation is didactic. For real work, reach for a dedicated matrix-completion
library rather than hand-rolling the loop.

```python
np.corrcoef(Xapp[ismiss], X[ismiss])[0, 1]
```

## 12.5.3 Clustering

### K-Means Clustering

We start with simulated data that genuinely has two clusters, so we can check whether
K-means recovers something we know is there. The first 25 observations are shifted +3 on
the first coordinate and −4 on the second; the remaining 25 are left alone.

```python
np.random.seed(0)
X = np.random.standard_normal((50, 2))
X[:25, 0] += 3
X[:25, 1] -= 4
```

K-means with K = 2. `random_state` fixes the random initialization so the result is
reproducible; `n_init=20` runs the algorithm from 20 different random starts and keeps the
best.

```python
kmeans = KMeans(n_clusters=2, random_state=2, n_init=20).fit(X)
```

`labels_` holds the cluster assignment for each of the 50 observations. Compare them with
the true split at index 25 the separation is essentially perfect, even though we never
told `KMeans()` anything about the groups.

The cluster *numbers* themselves are arbitrary: swapping the labels 0 and 1 describes the
same clustering.

```python
kmeans.labels_
```

Plotting the observations, colored by cluster assignment. With only two variables we can
plot the raw data directly; with more, you would run PCA first and plot the first two score
vectors instead.

```python
fig, ax = plt.subplots(1, 1, figsize=(8, 8))
ax.scatter(X[:, 0], X[:, 1], c=kmeans.labels_)
ax.set_title("K-Means Clustering Results with K=2")
```

In real data we do not know the true K. Here is what happens with K = 3: K-means has to
return three clusters, so it splits one of the two real groups. K-means will always give
you K clusters whether they mean anything is a separate question.

```python
kmeans = KMeans(n_clusters=3, random_state=3, n_init=20).fit(X)
fig, ax = plt.subplots(figsize=(8, 8))
ax.scatter(X[:, 0], X[:, 1], c=kmeans.labels_)
ax.set_title("K-Means Clustering Results with K=3")
```

`n_init` matters. `inertia_` is the total within-cluster sum of squares, the quantity
K-means is minimizing (12.17), so lower is better. With a single random start we land on a
worse local optimum (78.06) than with 20 starts (75.04).

Two practical rules: always run with a large `n_init` (20 or 50), and always set
`random_state` so the run is reproducible.

```python
kmeans1 = KMeans(n_clusters=3, random_state=3, n_init=1).fit(X)
kmeans20 = KMeans(n_clusters=3, random_state=3, n_init=20).fit(X)
kmeans1.inertia_, kmeans20.inertia_
```

### Hierarchical Clustering

`AgglomerativeClustering` implements hierarchical clustering. We alias it to `HClust` for
brevity — this does not change the class of the objects it creates.

Passing `distance_threshold=0` together with `n_clusters=None` tells it to build the
*entire* dendrogram rather than stopping at a fixed number of clusters.

We start with **complete linkage** and Euclidean distance, on the same simulated `X` from
the K-means section.

```python
HClust = AgglomerativeClustering
hc_comp = HClust(distance_threshold=0, n_clusters=None, linkage='complete')
hc_comp.fit(X)
```

The same fit with **average** and **single** linkage. Linkage determines how the
dissimilarity between two *clusters* is derived from the pairwise dissimilarities of their
members: the maximum (complete), the mean (average), or the minimum (single).

```python
hc_avg = HClust(distance_threshold=0, n_clusters=None, linkage='average')
hc_avg.fit(X)
hc_sing = HClust(distance_threshold=0, n_clusters=None, linkage='single')
hc_sing.fit(X)
```

You can also supply your own dissimilarity matrix instead of letting scikit-learn compute
distances. The loop below builds the 50 × 50 matrix of pairwise Euclidean distances, and
`metric='precomputed'` tells `HClust` to treat the input as distances rather than
observations.

This is the route to take whenever your notion of dissimilarity is not one scikit-learn
knows how to compute.

```python
D = np.zeros((X.shape[0], X.shape[0]))
for i in range(X.shape[0]):
    x_ = np.multiply.outer(np.ones(X.shape[0]), X[i])
    D[i] = np.sqrt(np.sum((X - x_)**2, 1))
hc_sing_pre = HClust(distance_threshold=0, n_clusters=None,
                     metric='precomputed', linkage='single')
hc_sing_pre.fit(D)
```

To draw the tree we use SciPy's `dendrogram()`, but it expects a linkage matrix, which
`AgglomerativeClustering` does not provide hence `compute_linkage()` from `ISLP.cluster`.

By default `dendrogram()` colors branches according to a pre-chosen cut of the tree. We
override that by setting `color_threshold` to `-inf` and coloring everything above it
black. Since we want this for several plots, the settings live in the `cargs` dictionary
and get passed with `**cargs`.

The numbers along the bottom identify the observations. Height on the vertical axis is the
dissimilarity at which two clusters were fused: the higher the fusion, the less similar the
groups.

```python
cargs = {'color_threshold': -np.inf, 'above_threshold_color': 'black'}
linkage_comp = compute_linkage(hc_comp)
fig, ax = plt.subplots(1, 1, figsize=(8, 8))
dendrogram(linkage_comp, ax=ax, **cargs)
```

Setting `color_threshold=4` colors the branches that merge *below* height 4 differently
from those above it, which makes a particular cut of the tree visible at a glance.

```python
fig, ax = plt.subplots(1, 1, figsize=(8, 8))
dendrogram(linkage_comp, ax=ax, color_threshold=4,
           above_threshold_color='black')
```

`cut_tree()` converts a cut of the dendrogram into cluster labels. Here we cut so that four
clusters result.

The alternative passing `n_clusters` to `HClust()` would force a refit for every cut.
Cutting a linkage matrix is free once the tree is built.

```python
cut_tree(linkage_comp, n_clusters=4).T
```

You can also cut at a given *height* instead of asking for a number of clusters. Cutting at
height 5 happens to leave three clusters here.

```python
cut_tree(linkage_comp, height=5)
```

Just as with PCA, scaling the variables before clustering changes the result, because it
changes the distances between observations. The same `StandardScaler` applies.

```python
scaler = StandardScaler()
X_scale = scaler.fit_transform(X)
hc_comp_scale = HClust(distance_threshold=0, n_clusters=None,
                       linkage='complete').fit(X_scale)
linkage_comp_scale = compute_linkage(hc_comp_scale)
fig, ax = plt.subplots(1, 1, figsize=(8, 8))
dendrogram(linkage_comp_scale, ax=ax, **cargs)
ax.set_title("Hierarchical Clustering with Scaled Features")
```

**Correlation-based dissimilarity.** Instead of Euclidean distance, we can measure how
similar two observations are by the correlation across their feature values, and use
`1 - correlation` as the distance. Two observations with similar *profiles* across features
are then close, even if they differ in overall magnitude.

This only makes sense with at least three features: with two features, the absolute
correlation between any two observations is always exactly 1.

```python
X = np.random.standard_normal((30, 3))
corD = 1 - np.corrcoef(X)
hc_cor = HClust(linkage='complete', distance_threshold=0,
                n_clusters=None, metric='precomputed')
hc_cor.fit(corD)
linkage_cor = compute_linkage(hc_cor)
fig, ax = plt.subplots(1, 1, figsize=(8, 8))
dendrogram(linkage_cor, ax=ax, **cargs)
ax.set_title("Complete Linkage with Correlation-Based Dissimilarity")
```

## 12.5.4 NCI60 Data Example

The `NCI60` data is a classic genomics application: 6,830 gene expression measurements on
64 cancer cell lines.

Each cell line carries a cancer-type label. We do **not** use those labels when fitting
PCA and clustering are unsupervised. We hold them back and check afterwards how well the
unsupervised structure lines up with the known types.

```python
NCI60 = load_data('NCI60')
nci_labs = NCI60['labels']
nci_data = NCI60['data']
```

64 rows (cell lines) and 6,830 columns (genes) far more variables than observations,
which is typical of genomic data and exactly the setting where dimension reduction earns
its keep.

```python
nci_data.shape
```

The distribution of cancer types across the cell lines. Several types have only a handful
of cell lines, and a few entries are repeated measurements of the same line (the `-repro`
labels).

```python
nci_labs.value_counts()
```

### PCA on the NCI60 Data

We scale the genes to standard deviation one before PCA. This is a judgement call rather
than a necessity here: unlike `USArrests`, the genes are all measured in the same units, so
there is a reasonable argument for leaving them unscaled.

`fit_transform()` fits and returns the scores in one step.

```python
scaler = StandardScaler()
nci_scaled = scaler.fit_transform(nci_data)
nci_pca = PCA()
nci_scores = nci_pca.fit_transform(nci_scaled)
```

Plotting the cell lines in the space of the first few principal components, colored by
cancer type. If cell lines of the same type have similar expression profiles, they should
land near each other — and broadly they do, though not cleanly.

Without a dimension reduction method there would be no way to look at this data at all:
with 6,830 genes there are millions of possible pairwise scatterplots, and none of them
would be especially informative on its own.

```python
cancer_types = list(np.unique(nci_labs))
nci_groups = np.array([cancer_types.index(lab) for lab in nci_labs.values])
fig, axes = plt.subplots(1, 2, figsize=(15, 6))
ax = axes[0]
ax.scatter(nci_scores[:, 0], nci_scores[:, 1], c=nci_groups, marker='o', s=50)
ax.set_xlabel('PC1'); ax.set_ylabel('PC2')
ax = axes[1]
ax.scatter(nci_scores[:, 0], nci_scores[:, 2], c=nci_groups, marker='o', s=50)
ax.set_xlabel('PC1'); ax.set_ylabel('PC3')
```

The PVE and cumulative PVE for the NCI60 components.

The first seven components together explain only about 40% of the variance not a lot.
But the scree plot shows a clear elbow around the seventh component: the components after
that each contribute very little, so there is probably little to gain from examining more
than seven or so. (Seven is already hard enough to interpret.)

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 6))
ax = axes[0]
ticks = np.arange(nci_pca.n_components_)+1
ax.plot(ticks, nci_pca.explained_variance_ratio_, marker='o')
ax.set_xlabel('Principal Component')
ax.set_ylabel('PVE')
ax = axes[1]
ax.plot(ticks, nci_pca.explained_variance_ratio_.cumsum(), marker='o')
ax.set_xlabel('Principal Component')
ax.set_ylabel('Cumulative PVE')
```

### Clustering the Observations of the NCI60 Data

A small helper so we can produce the three dendrograms without repeating ourselves. It
fits the hierarchical clustering with the requested linkage, builds the linkage matrix,
draws the dendrogram with the cell lines labelled by cancer type, and returns the fitted
object.

The `cut` argument lets us later draw a specific cut of the tree in a different color.

```python
def plot_nci(linkage, ax, cut=-np.inf):
    cargs = {'above_threshold_color': 'black', 'color_threshold': cut}
    hc = HClust(n_clusters=None, distance_threshold=0,
                linkage=linkage.lower()).fit(nci_scaled)
    linkage_ = compute_linkage(hc)
    dendrogram(linkage_, ax=ax, labels=np.asarray(nci_labs),
               leaf_font_size=10, **cargs)
    ax.set_title('%s Linkage' % linkage)
    return hc
```

The three linkages side by side. The choice clearly matters:

- **Single linkage** produces *trailing* clusters one large cluster onto which individual
  observations attach one at a time.
- **Complete** and **average** linkage give more balanced, evenly sized clusters, which is
  why they are generally preferred.

Cell lines of the same cancer type do tend to cluster together, though far from perfectly.
We use complete linkage for the rest of the analysis.

```python
fig, axes = plt.subplots(3, 1, figsize=(15, 30))
ax = axes[0]; hc_comp = plot_nci('Complete', ax)
ax = axes[1]; hc_avg = plot_nci('Average', ax)
ax = axes[2]; hc_sing = plot_nci('Single', ax)
```

Cutting the complete-linkage tree into four clusters and cross-tabulating against the true
cancer types shows some clear structure: all the leukemia cell lines land in a single
cluster, while the breast cancer lines are scattered across three different clusters.

```python
linkage_comp = compute_linkage(hc_comp)
comp_cut = cut_tree(linkage_comp, n_clusters=4).reshape(-1)
pd.crosstab(nci_labs['label'],
            pd.Series(comp_cut.reshape(-1), name='Complete'))
```

The same four-cluster cut, drawn on the dendrogram. `axhline()` draws a horizontal line on
the existing axes; height 140 is where the tree happens to split into four groups. The
clusters this line produces are the same ones stored in `comp_cut`.

```python
fig, ax = plt.subplots(figsize=(10, 10))
plot_nci('Complete', ax, cut=140)
ax.axhline(140, c='r', linewidth=4)
```

Section 12.4.2 warned that K-means and a dendrogram cut to the same number of clusters can
disagree. Here is the comparison with K = 4.

Remember that the cluster labels are arbitrary in both methods, so read the table by
looking for rows and columns that line up. K-means cluster 3 is identical to hierarchical
cluster 2. The rest diverge: K-means cluster 0 absorbs part of hierarchical cluster 0 plus
all of hierarchical cluster 1.

```python
nci_kmeans = KMeans(n_clusters=4, random_state=0, n_init=20).fit(nci_scaled)
pd.crosstab(pd.Series(comp_cut, name='HClust'),
            pd.Series(nci_kmeans.labels_, name='K-means'))
```

A useful variation: rather than clustering the full 6,830-column matrix, cluster the first
five principal component score vectors. Those components carry the dominant signal, so this
amounts to clustering a denoised version of the data and it sometimes gives cleaner
results than clustering the raw measurements.

```python
hc_pca = HClust(n_clusters=None, distance_threshold=0,
                linkage='complete').fit(nci_scores[:, :5])
linkage_pca = compute_linkage(hc_pca)
fig, ax = plt.subplots(figsize=(8, 8))
dendrogram(linkage_pca, labels=np.asarray(nci_labs),
           leaf_font_size=10, ax=ax, **cargs)
ax.set_title("Hier. Clust. on First Five Score Vectors")
pca_labels = pd.Series(cut_tree(linkage_pca, n_clusters=4).reshape(-1),
                       name='Complete-PCA')
pd.crosstab(nci_labs['label'], pca_labels)
```

| 12.5.1 | PCA on `USArrests` | Standardize when variables have different scales, or the largest-variance variable dominates. Loadings live in `components_`, scores come from `transform()`, PVE from `explained_variance_ratio_`. Component signs are arbitrary. |
| 12.5.2 | Matrix completion via SVD | Iteratively: fill missing entries, take a rank-M approximation, re-fill, repeat until the relative error stops improving. Recovered the 20 hidden values with correlation ≈ 0.71. |
| 12.5.3 | K-means | Needs K up front and will always return K clusters. `inertia_` is the objective being minimized; use a large `n_init` and a fixed `random_state`. |
| 12.5.3 | Hierarchical clustering | `distance_threshold=0, n_clusters=None` builds the full tree; `compute_linkage()` + `dendrogram()` draws it; `cut_tree()` cuts it by count or height. Linkage choice and scaling both change the answer. |
| 12.5.4 | `NCI60` | PCA makes a 6,830-dimension dataset visible; complete-linkage clustering roughly recovers cancer types (leukemia cleanly, breast cancer not). K-means and hierarchical clustering at K = 4 agree only partially. |

**The caution from Section 12.5 is worth keeping in mind:** clustering results are sensitive to small choices how the data are standardized, which linkage is used, where the tree is cut and are not robust to perturbing the data. Vary these choices, see which patterns
persist, and treat the output as the starting point for a hypothesis rather than the truth about the dataset.
