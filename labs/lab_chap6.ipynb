## 6.5.1 Subset Selection Methods

We start by loading the `Hitters` data. The `Salary` column has missing values (59 players). We drop those rows to have a clean dataset for modeling.

```python
Hitters = load_data('Hitters')
np.isnan(Hitters['Salary']).sum()
```

```python
Hitters = Hitters.dropna()
Hitters.shape
```

### Forward Selection using Mallows' Cp

We want to select the best subset of predictors using forward stepwise selection. Instead of using MSE (which always improves with more variables), we use **Mallows' Cp** as a criterion to penalize model complexity.

Since `sklearn` maximizes scores, we define a **negative Cp** function so that a smaller Cp becomes a larger (better) score.

```python
def nCp(sigma2, estimator, X, Y):
    "Negative Cp statistic"
    n, p = X.shape
    Yhat = estimator.predict(X)
    RSS = np.sum((Y - Yhat)**2)
    return -(RSS + 2 * p * sigma2) / n
```

We create the design matrix (all predictors except `Salary`) and estimate the residual variance $\sigma^2$ from the full (least squares) model. This $\sigma^2$ is needed in the Cp formula.

```python
design = MS(Hitters.columns.drop('Salary')).fit(Hitters)
Y = np.array(Hitters['Salary'])
X = design.transform(Hitters)
sigma2 = OLS(Y, X).fit().scale
```

```python
neg_Cp = partial(nCp, sigma2)
```

We define a forward stepwise strategy that continues adding variables until adding another variable **does not improve** the scoring criterion. This is the `first_peak` method.

```python
strategy = Stepwise.first_peak(design, direction='forward', max_terms=len(design.terms))
```

First, we use **MSE** as the selection criterion (default). This will likely select all variables because training MSE always decreases when adding variables.

Then, we use **negative Cp** (our custom scorer). The Cp criterion adds a penalty for complexity, so we expect a smaller model.

```python
hitters_MSE = sklearn_selected(OLS, strategy)
hitters_MSE.fit(Hitters, Y)
hitters_MSE.selected_state_
```

```python
hitters_Cp = sklearn_selected(OLS, strategy, scoring=neg_Cp)
hitters_Cp.fit(Hitters, Y)
hitters_Cp.selected_state_
```

**Interpretation**: With Cp, the forward selection chooses only **10 variables**, whereas using MSE would have included all 19. This shows how Cp penalizes overfitting.

### Model Selection using Validation Set & Cross-Validation

Instead of relying on Cp, we can use **cross-validation** to choose the number of steps in forward selection. For this, we need to store the **entire path** of models (from 0 to 19 variables). The `sklearn_selection_path` function does this.

```python
strategy = Stepwise.fixed_steps(design, len(design.terms), direction='forward')
full_path = sklearn_selection_path(OLS, strategy)
```

Fit the full path and compute **in-sample MSE**. As expected, the in-sample MSE decreases monotonically as we add more variables – this is why we cannot use it to select the best model size.

```python
full_path.fit(Hitters, Y)
Yhat_in = full_path.predict(Hitters)
Yhat_in.shape
```

```python
mse_fig, ax = subplots(figsize=(8,8))
insample_mse = ((Yhat_in - Y[:, None])**2).mean(0)
n_steps = insample_mse.shape[0]
ax.plot(np.arange(n_steps), insample_mse, 'k', label='In-sample')
ax.set_ylabel('MSE', fontsize=20)
ax.set_xlabel('Steps of forward stepwise', fontsize=20)
ax.set_xticks(np.arange(n_steps)[::2])
ax.legend()
ax.set_ylim([50000,250000]);
```

We now compute **5-fold cross-validated predictions** for each model along the path. The `cross_val_predict` function returns predictions for each observation using models trained on folds that excluded that observation.

```python
K=5
kfold=skm.KFold(K, random_state=0, shuffle=True)
Yhat_cv=skm.cross_val_predict(full_path, Hitters, Y, cv=kfold)
Yhat_cv.shape
```

We manually compute the MSE for each of the 5 folds and average them to get the cross-validated MSE for each step.

```python
cv_mse=[]
for train_idx,test_idx in kfold.split(Y):
    errors=(Yhat_cv[test_idx]- Y[test_idx,None])**2
    cv_mse.append(errors.mean(0))
cv_mse=np.array(cv_mse).T
cv_mse.shape
```

```python
ax.errorbar(np.arange(n_steps), cv_mse.mean(1), cv_mse.std(1) / np.sqrt(K), label='Cross-validated', c='r')
ax.set_ylim([50000,250000])
ax.legend()
mse_fig
```

**Interpretation**: The cross-validated MSE initially drops and then starts to level off or increase, indicating that we should stop before adding too many variables. The validation set approach (a single 80/20 split) gives a similar idea – we try that next.

```python
validation = skm.ShuffleSplit(n_splits=1, test_size=0.2, random_state=0)
for train_idx, test_idx in validation.split(Y):
    full_path.fit(Hitters.iloc[train_idx], Y[train_idx])
    Yhat_val = full_path.predict(Hitters.iloc[test_idx])
    errors = (Yhat_val - Y[test_idx,None])**2
    validation_mse = errors.mean(0)
```

```python
ax.plot(np.arange(n_steps), validation_mse, 'b--', label='Validation')
ax.set_xticks(np.arange(n_steps)[::2])
ax.set_ylim([50000,250000])
ax.legend()
mse_fig
```

## 6.5.2 Ridge Regression and the Lasso

We now turn to **regularized linear models**. Both ridge and lasso add a penalty to the RSS. The key difference:
- **Ridge** ($\ell_2$ penalty) shrinks coefficients but keeps all variables.
- **Lasso** ($\ell_1$ penalty) can shrink coefficients to exactly zero, performing variable selection.

We use `sklearn.linear_model.ElasticNet` (with `l1_ratio=0` for ridge, `l1_ratio=1` for lasso).

### Ridge Regression

First, we manually standardize the predictors (important because ridge is sensitive to scale) and compute a **solution path** over a grid of $\lambda$ values.

```python
Xs = X - X.mean(0)[None,:]
X_scale = X.std(0)
Xs = Xs / X_scale[None,:]
lambdas = 10**np.linspace(8, -2, 100) / Y.std()
soln_array = skl.ElasticNet.path(Xs, Y, l1_ratio=0., alphas=lambdas)[1]
soln_array.shape
```

We store the coefficients in a DataFrame indexed by $-\log(\lambda)$ (so that larger values mean less shrinkage, moving left to right on the plot).

```python
soln_path = pd.DataFrame(soln_array.T, columns=D.columns, index=-np.log(lambdas))
soln_path.index.name = 'negative log(lambdas)'
soln_path
```

Plotting the ridge coefficients as a function of $-\log(\lambda)$: when $\lambda$ is large (left side of plot), coefficients are heavily shrunk toward zero. When $\lambda$ is small (right side), coefficients approach the least squares estimates.

```python
path_fig, ax = subplots(figsize=(8,8))
soln_path.plot(ax=ax, legend=False)
ax.set_xlabel('$-\log(\lambda)$', fontsize=20)
ax.set_ylabel('Standardized coefficients', fontsize=20)
ax.legend(loc='upper left');
```

Let's check the $\ell_2$ norm of the coefficients at two different $\lambda$ values. For a large $\lambda$ (25.5), the norm is small (24.2). For a small $\lambda$ (0.244), the norm is much larger (160.4), demonstrating that smaller $\lambda$ leads to less shrinkage.

```python
beta_hat = soln_path.loc[soln_path.index[39]]
lambdas[39], beta_hat
```

```python
np.linalg.norm(beta_hat)
```

```python
beta_hat = soln_path.loc[soln_path.index[59]]
lambdas[59], np.linalg.norm(beta_hat)
```

A more robust way is to use a `Pipeline` that automatically standardizes the features before fitting the ridge model. This ensures that standardization is part of the model object, which is crucial for cross-validation.

```python
ridge = skl.ElasticNet(alpha=lambdas[59], l1_ratio=0)
scaler = StandardScaler(with_mean=True, with_std=True)
pipe = Pipeline(steps=[('scaler', scaler), ('ridge', ridge)])
pipe.fit(X, Y)
```

```python
np.linalg.norm(ridge.coef_)
```

### Estimating Test Error of Ridge Regression

We need to choose a good value for $\lambda$. We can use **validation set** or **cross-validation**. Here we use `GridSearchCV` over our grid of `lambdas`.

First, using a 50% validation split.

```python
validation = skm.ShuffleSplit(n_splits=1, test_size=0.5, random_state=0)
ridge.alpha = 0.01
results = skm.cross_validate(ridge, X, Y, scoring='neg_mean_squared_error', cv=validation)
-results['test_score']
```

```python
ridge.alpha = 1e10
results = skm.cross_validate(ridge, X, Y, scoring='neg_mean_squared_error', cv=validation)
-results['test_score']
```

Now we use `GridSearchCV` to systematically search over `lambdas` using the validation set (and later 5-fold CV).

```python
param_grid = {'ridge__alpha': lambdas}
grid = skm.GridSearchCV(pipe, param_grid, cv=validation, scoring='neg_mean_squared_error')
grid.fit(X, Y)
grid.best_params_['ridge__alpha']
grid.best_estimator_
```

```python
grid = skm.GridSearchCV(pipe, param_grid, cv=kfold, scoring='neg_mean_squared_error')
grid.fit(X, Y)
grid.best_params_['ridge__alpha']
grid.best_estimator_
```

We plot the cross-validated MSE against $-\log(\lambda)$. The minimum MSE is achieved at an intermediate value of $\lambda$ – not too large (over-shrink) and not too small (overfit).

```python
ridge_fig, ax = subplots(figsize=(8,8))
ax.errorbar(-np.log(lambdas),
            -grid.cv_results_['mean_test_score'],
            yerr=grid.cv_results_['std_test_score'] / np.sqrt(K))
ax.set_ylim([50000,250000])
ax.set_xlabel('$-\log(\lambda)$', fontsize=20)
ax.set_ylabel('Cross-validated MSE', fontsize=20);
```

We can also use **$R^2$** as the scoring metric. The shape of the curve will be similar but inverted.

```python
grid_r2 = skm.GridSearchCV(pipe, param_grid, cv=kfold)
grid_r2.fit(X, Y)
```

```python
r2_fig, ax = subplots(figsize=(8,8))
ax.errorbar(-np.log(lambdas),
            grid_r2.cv_results_['mean_test_score'],
            yerr=grid_r2.cv_results_['std_test_score'] / np.sqrt(K))
ax.set_xlabel('$-\log(\lambda)$', fontsize=20)
ax.set_ylabel('Cross-validated $R^2$', fontsize=20);
```

### Fast Cross-Validation for Solution Paths

Because ridge and lasso have efficient algorithms to fit the entire path, `sklearn` provides `ElasticNetCV` which does cross-validation over the path more efficiently. Note: the results may differ slightly from `GridSearchCV` because standardization is done only once globally in this pipeline, whereas `GridSearchCV` standardizes per fold (more correct, but often similar).

```python
ridgeCV = skl.ElasticNetCV(alphas=lambdas, l1_ratio=0, cv=kfold)
pipeCV = Pipeline(steps=[('scaler', scaler), ('ridge', ridgeCV)])
pipeCV.fit(X, Y)
```

```python
tuned_ridge = pipeCV.named_steps['ridge']
ridgeCV_fig, ax = subplots(figsize=(8,8))
ax.errorbar(-np.log(lambdas), tuned_ridge.mse_path_.mean(1),
            yerr=tuned_ridge.mse_path_.std(1) / np.sqrt(K))
ax.axvline(-np.log(tuned_ridge.alpha_), c='k', ls='--')
ax.set_ylim([50000,250000])
ax.set_xlabel('$-\log(\lambda)$', fontsize=20)
ax.set_ylabel('Cross-validated MSE', fontsize=20);
```

The optimal $\lambda$ is printed as `tuned_ridge.alpha_`. The minimum cross-validated MSE is about **115,527**, better than the null model and least squares.

```python
np.min(tuned_ridge.mse_path_.mean(1))
```

```python
tuned_ridge.coef_
```

### Evaluating Test Error of Cross-Validated Ridge

If we want an honest estimate of the test error for the *tuned* ridge model, we must use a **nested** procedure: split the data into an outer test set and an inner training set. We use the training set for cross-validation (to choose $\lambda$), then evaluate on the test set.

```python
outer_valid = skm.ShuffleSplit(n_splits=1, test_size=0.25, random_state=1)
inner_cv = skm.KFold(n_splits=5, shuffle=True, random_state=2)
ridgeCV = skl.ElasticNetCV(alphas=lambdas, l1_ratio=0, cv=inner_cv)
pipeCV = Pipeline(steps=[('scaler', scaler), ('ridge', ridgeCV)])
results = skm.cross_validate(pipeCV, X, Y, cv=outer_valid, scoring='neg_mean_squared_error')
-results['test_score']
```

### The Lasso

Now we fit the **lasso** ($\ell_1$ penalty) using the same `ElasticNetCV` but with `l1_ratio=1`. The lasso has the advantage of producing sparse models – some coefficients are exactly zero.

We fit the path and plot the coefficients.

```python
lassoCV = skl.ElasticNetCV(alphas=100, l1_ratio=1, cv=kfold)
pipeCV = Pipeline(steps=[('scaler', scaler), ('lasso', lassoCV)])
pipeCV.fit(X, Y)
tuned_lasso = pipeCV.named_steps['lasso']
tuned_lasso.alpha_
```

```python
lambdas, soln_array = skl.Lasso.path(Xs, Y, n_alphas=100)[:2]
soln_path = pd.DataFrame(soln_array.T, columns=D.columns, index=-np.log(lambdas))
```

```python
path_fig, ax = subplots(figsize=(8,8))
soln_path.plot(ax=ax, legend=False)
ax.legend(loc='upper left')
ax.set_xlabel('$-\log(\lambda)$', fontsize=20)
ax.set_ylabel('Standardized coefficients', fontsize=20);
```

The minimum cross-validated MSE for the lasso is about **114,691**, slightly better than ridge. More importantly, the lasso model includes only 13 variables (6 coefficients are zero), making it more interpretable.

```python
np.min(tuned_lasso.mse_path_.mean(1))
```

```python
lassoCV_fig, ax = subplots(figsize=(8,8))
ax.errorbar(-np.log(tuned_lasso.alphas_),
            tuned_lasso.mse_path_.mean(1),
            yerr=tuned_lasso.mse_path_.std(1) / np.sqrt(K))
ax.axvline(-np.log(tuned_lasso.alpha_), c='k', ls='--')
ax.set_ylim([50000,250000])
ax.set_xlabel('$-\log(\lambda)$', fontsize=20)
ax.set_ylabel('Cross-validated MSE', fontsize=20);
```

## 6.5.3 PCR and PLS Regression

### Principal Components Regression (PCR)

PCR transforms the predictors into principal components (unsupervised) and then uses the first $M$ components in a linear regression. Because PCA is sensitive to scaling, we standardize the predictors.

First, we fit a PCR with 2 components.

```python
pca = PCA(n_components=2)
linreg = skl.LinearRegression()
pipe = Pipeline([('pca', pca), ('linreg', linreg)])
pipe.fit(X, Y)
pipe.named_steps['linreg'].coef_
```

```python
pipe = Pipeline([('scaler', scaler), ('pca', pca), ('linreg', linreg)])
pipe.fit(X, Y)
pipe.named_steps['linreg'].coef_
```

We use cross-validation to choose the optimal number of principal components. The grid searches from 1 to 19 components.

```python
param_grid = {'pca__n_components': range(1, 20)}
grid = skm.GridSearchCV(pipe, param_grid, cv=kfold, scoring='neg_mean_squared_error')
grid.fit(X, Y)
```

```python
pcr_fig, ax = subplots(figsize=(8,8))
n_comp = param_grid['pca__n_components']
ax.errorbar(n_comp,
            -grid.cv_results_['mean_test_score'],
            grid.cv_results_['std_test_score'] / np.sqrt(K))
ax.set_ylabel('Cross-validated MSE', fontsize=20)
ax.set_xlabel('Principal components', fontsize=20)
ax.set_xticks(n_comp[::2])
ax.set_ylim([50000,250000]);
```

**Interpretation**: The lowest CV error occurs with 17 components, but the curve is fairly flat – using just a few components (e.g., 1 or 2) gives nearly the same performance. This suggests that a low-dimensional projection captures most of the predictive information.

Finally, we compute the cross-validated MSE for the **null model** (intercept only) to compare.

```python
Xn = np.zeros((X.shape[0], 1))
cv_null = skm.cross_validate(linreg, Xn, Y, cv=kfold, scoring='neg_mean_squared_error')
-cv_null['test_score']
```
