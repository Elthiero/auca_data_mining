# ISLP Chapter 8 Lab: Tree-Based Methods

**Text:** *An Introduction to Statistical Learning with Applications in Python*, 8.3

A decision tree splits the feature space into rectangles by recursive binary splitting, then
predicts a constant in each. A single tree is highly interpretable but has high variance
a different training sample can produce a very different tree. The rest of the chapter is about
trading interpretability for accuracy by averaging many trees:

| Section | Method | How the trees are grown |
|---|---|---|
| 8.3.1 | Classification tree | One tree, pruned by cost-complexity |
| 8.3.2 | Regression tree | One tree, same pruning idea, squared-error splits |
| 8.3.3 | Bagging | Many trees, independent, on bootstrap samples |
| 8.3.3 | Random forest | As bagging, but each split sees a random subset of features |
| 8.3.4 | Boosting | Trees grown sequentially on the residual, shrunken |
| 8.3.5 | BART | Trees grown sequentially, each perturbed via MCMC |

Two datasets: `Carseats` (classification, `Sales > 8`) and `Boston` (regression, `medv`).

**Requirements:** `ISLP`, `scikit-learn`, `pandas`, `numpy`, `matplotlib`, `statsmodels`.

The new imports for this lab. `DTC`/`DTR` are the single-tree estimators, `RF` serves for

## 8.3.1 Fitting Classification Trees

### Recoding the response

`Carseats` records unit sales at 400 stores. `Sales` is continuous, so it is recoded into a
binary label: `High = "Yes"` when sales exceed 8. `Sales` itself must then be dropped from
the predictors keeping it would leak the answer, since `High` is a deterministic function
of it.

```python
Carseats = load_data('Carseats')
High = np.where(Carseats.Sales > 8,
                "Yes",
                "No")

Carseats.head()
```

### Building the model matrix

`ModelSpec` one-hot-encodes the qualitative predictors (`ShelveLoc`, `Urban`, `US`).
`intercept=False` because a tree has no intercept to estimate.

Worth knowing: trees do not *need* dummy variables. A tree can split a categorical predictor
by partitioning its levels into two groups directly, which is more natural and often produces
shallower trees. scikit-learn does not implement that, so each dummy column is treated as its
own numeric variable and splits look like `ShelveLoc[Good] <= 0.5`. `feature_names` is kept
for annotating the plots.

```python
model = MS(Carseats.columns.drop('Sales'), intercept=False)
D = model.fit_transform(Carseats)
feature_names = list(D.columns)
X = np.asarray(D)

X.shape
```

### Fitting a shallow tree

The parameters that matter most are `max_depth` (how deep to grow), `min_samples_split`
(how many observations a node needs before it can be split) and `criterion` `'entropy'`
for cross-entropy or `'gini'` for the Gini index. Both measure node purity and usually give
similar trees; classification error rate is too blunt for growing, though it is used for pruning.

`random_state` is set because ties in the split criterion are broken at random.

```python
clf = DTC(criterion='entropy',
          max_depth=3,
          random_state=0)
clf.fit(X, High)
```

### Training accuracy and deviance

Accuracy is about 0.7275, i.e. a **training** error rate of roughly 27.5% the book's text
says 21% here, but the printed output is 0.7275, so 27.5% is what this fit gives. That is
training error either way, and it is optimistic; the honest number comes from the held-out
evaluation two cells further down.

```python
accuracy_score(High, clf.predict(X))
```

For classification trees, the deviance is

$$-2\sum_m \sum_k n_{mk}\log \hat p_{mk},$$

where $n_{mk}$ is the number of observations in terminal node $m$ belonging to class $k$.
`log_loss` computes the cross-entropy version of this. Smaller means a closer fit to the
training data and, as always, a smaller training deviance alone is not evidence of a
better model.

```python
resid_dev = np.sum(log_loss(High, clf.predict_proba(X)))
resid_dev
```

### Displaying the tree

The chief attraction of a single tree is that it can be read. Each node shows the split rule,
the impurity, the number of samples reaching it, and the class counts.

`ShelveLoc[Good]` is the root split shelf location is the strongest single indicator of high
sales, which is a plausible retail finding.

```python
ax = subplots(figsize=(12, 12))[1]
plot_tree(clf,
          feature_names=feature_names,
          ax=ax);
```

The text version is often easier to quote in a report. `show_weights=True` prints the
`[No, Yes]` counts in each leaf, so the confidence behind each prediction is visible  a leaf
reading `[7, 29]` is on much firmer ground than one reading `[5, 6]`.

```python
print(export_text(clf,
                  feature_names=feature_names,
                  show_weights=True))
```

### Estimating test error properly

Training accuracy tells us nothing about generalization. `ShuffleSplit` with `n_splits=1`
holds out 200 observations once a plain validation-set split expressed through the
cross-validation API. The same pattern works for any estimator; only the model object changes.

The result is around 0.685, noticeably below the 0.7275 training figure.

```python
validation = skm.ShuffleSplit(n_splits=1,
                              test_size=200,
                              random_state=0)
results = skm.cross_validate(clf,
                             D,
                             High,
                             cv=validation)
results['test_score']
```

### Pruning by cost-complexity

Growing a deep tree and then pruning back beats stopping early, because a weak split can lead
to a strong one further down. Cost-complexity pruning minimizes

$$\sum_{m=1}^{|T|} \sum_{x_i \in R_m} (y_i - \hat y_{R_m})^2 + \alpha |T|,$$

where $|T|$ is the number of terminal nodes. Increasing $\alpha$ buys smaller trees; $\alpha$
is chosen by cross-validation.

First, a proper split: half for training (grow and prune here), half held back for the final
evaluation.

```python
(X_train,
 X_test,
 High_train,
 High_test) = skm.train_test_split(X,
                                   High,
                                   test_size=0.5,
                                   random_state=0)
```

The full tree is grown with no depth limit the size will be learned by cross-validation
rather than fixed by hand. Its test accuracy is about 0.735, the benchmark the pruned tree
must beat.

```python
clf = DTC(criterion='entropy', random_state=0)
clf.fit(X_train, High_train)
accuracy_score(High_test, clf.predict(X_test))
```

`cost_complexity_pruning_path()` returns the finite set of $\alpha$ values at which the
optimal subtree changes everything in between gives the identical tree, so only these
need to be searched. That turns a continuous tuning problem into a small grid.

```python
ccp_path = clf.cost_complexity_pruning_path(X_train, High_train)
kfold = skm.KFold(10,
                  random_state=1,
                  shuffle=True)
```

`GridSearchCV` scores each $\alpha$ by 10-fold CV on the **training half only** the test
half is untouched, which is what keeps the final number honest. `refit=True` then retrains
the best tree on all the training data.

Best CV accuracy comes out around 0.685.

```python
grid = skm.GridSearchCV(clf,
                        {'ccp_alpha': ccp_path.ccp_alphas},
                        refit=True,
                        cv=kfold,
                        scoring='accuracy')
grid.fit(X_train, High_train)
grid.best_score_
```

```python
ax = subplots(figsize=(12, 12))[1]
best_ = grid.best_estimator_
plot_tree(best_,
          feature_names=feature_names,
          ax=ax);
```

Still a bushy tree. Rather than counting leaves off the figure, query the fitted object:
30 terminal nodes.

```python
best_.tree_.n_leaves
```

### The honest verdict

The pruned tree gets about 72.0% of test observations right *worse* than the 73.5% of the
unpruned tree. Cross-validation removed only 5 leaves and bought nothing here.

That is a genuinely useful result to report rather than hide. Cross-validation is unbiased for
model selection but has real variance; with 200 training observations and a single split, a
different seed can reverse the ranking. The confusion table also shows where the errors live:
recall on the `Yes` class is poor many high-sales stores are predicted `No`.

```python
print(accuracy_score(High_test,
                     best_.predict(X_test)))
confusion = confusion_table(best_.predict(X_test),
                            High_test)
confusion
```

## 8.3.2 Fitting Regression Trees

The mechanics are identical; only the split criterion changes from impurity to residual sum of
squares, and each leaf predicts the mean response of its training observations rather than a
class. The response is `medv`, median home value in $1,000s.

```python
Boston = load_data("Boston")
model = MS(Boston.columns.drop('medv'), intercept=False)
D = model.fit_transform(Boston)
feature_names = list(D.columns)
X = np.asarray(D)

X.shape
```

```python
(X_train,
 X_test,
 y_train,
 y_test) = skm.train_test_split(X,
                                Boston['medv'],
                                test_size=0.3,
                                random_state=0)
```

A depth-3 tree, kept shallow so the structure is readable. `lstat` (percentage of residents of
lower socioeconomic status) dominates the top splits, and lower `lstat` means more expensive
housing. Reading one path end to end: small homes (`rm < 6.8`) in low-status suburbs
(`lstat > 14.4`) with moderate crime (`crim > 5.8`) are predicted at about $12,042.

The step-function nature of the model is visible here every suburb landing in that leaf gets
the identical prediction, no matter how it differs on the other variables.

```python
reg = DTR(max_depth=3)
reg.fit(X_train, y_train)
ax = subplots(figsize=(12, 12))[1]
plot_tree(reg,
          feature_names=feature_names,
          ax=ax);
```

Pruning works the same way, with `neg_mean_squared_error` as the scoring rule. scikit-learn
always *maximizes* a score, so MSE is negated the least negative value is the best model.

```python
ccp_path = reg.cost_complexity_pruning_path(X_train, y_train)
kfold = skm.KFold(5,
                  shuffle=True,
                  random_state=10)
grid = skm.GridSearchCV(reg,
                        {'ccp_alpha': ccp_path.ccp_alphas},
                        refit=True,
                        cv=kfold,
                        scoring='neg_mean_squared_error')
G = grid.fit(X_train, y_train)
```

Test MSE lands near 28.07. The square root, $\sqrt{28.07} \approx 5.30$, is the interpretable
figure: predictions are typically within about $5,300 of the true median home value. Always
report RMSE alongside MSE it is in the units of the response, and MSE alone is hard to
reason about.

This 28.07 is the baseline every ensemble below is trying to beat.

```python
best_ = grid.best_estimator_
np.mean((y_test - best_.predict(X_test))**2)
```

```python
ax = subplots(figsize=(12, 12))[1]
plot_tree(G.best_estimator_,
          feature_names=feature_names,
          ax=ax);
```

## 8.3.3 Bagging and Random Forests

Bagging grows $B$ trees on bootstrap resamples and averages them. Averaging $B$ independent
quantities each with variance $\sigma^2$ gives variance $\sigma^2/B$, which is exactly the
weakness of a single tree being addressed deep unpruned trees have low bias and high
variance, so averaging keeps the bias and removes much of the variance.

The catch is that bootstrap samples overlap heavily, so the trees are correlated and the
variance reduction falls short of $1/B$. Random forests fix this by restricting each split to
a random subset of $m$ predictors, which stops every tree from leading with the same dominant
variable and decorrelates them.

`RandomForestRegressor` covers both: `max_features = p` is bagging.

```python
bag_boston = RF(max_features=X_train.shape[1], random_state=0)
bag_boston.fit(X_train, y_train)
```

Test MSE around 14.63 — roughly half that of the optimally pruned single tree. That is the
headline result of the chapter: a large accuracy gain, paid for with interpretability, since
there is no longer a single tree to draw.

The scatter of predicted against actual is worth a look: the points should hug the 45° line,
and the visible ceiling near 50 is the known censoring of `medv` at that value in this dataset.

```python
ax = subplots(figsize=(8, 8))[1]
y_hat_bag = bag_boston.predict(X_test)
ax.scatter(y_hat_bag, y_test)
ax.set_xlabel('Predicted medv')
ax.set_ylabel('Actual medv')
np.mean((y_test - y_hat_bag)**2)
```

Raising the forest from 100 to 500 trees barely moves the MSE (14.61). This is the expected
behaviour: bagging and random forests **cannot overfit** by adding trees more trees only
refine the average though too few will underfit. `n_estimators` is therefore a
compute budget, not a tuning parameter to be cross-validated carefully.

```python
bag_boston = RF(max_features=X_train.shape[1],
                n_estimators=500,
                random_state=0).fit(X_train, y_train)
y_hat_bag = bag_boston.predict(X_test)
np.mean((y_test - y_hat_bag)**2)
```

### A genuine random forest

Setting `max_features=6` (half of $p = 12$) decorrelates the trees. Defaults differ by task:
`RandomForestRegressor` defaults to $p$ (i.e. bagging), while `RandomForestClassifier` defaults
to $\sqrt{p}$.

Here the forest scores 20.04 **worse** than bagging's 14.63. Decorrelation is not guaranteed
to help: when a few predictors are genuinely dominant, forcing splits to ignore them costs more
in bias than it gains in variance reduction. `max_features` is a real tuning parameter and
should be cross-validated rather than assumed.

```python
RF_boston = RF(max_features=6,
               random_state=0).fit(X_train, y_train)
y_hat_RF = RF_boston.predict(X_test)
np.mean((y_test - y_hat_RF)**2)
```

### Variable importance

With no tree to plot, importance scores are how an ensemble is interpreted. Each is the total
decrease in node impurity attributable to splits on that variable, averaged over all trees.
`lstat` and `rm` dominate community wealth and house size.

One caveat worth stating in any writeup: impurity-based importance is biased toward
high-cardinality and continuous predictors, and it splits credit arbitrarily between correlated
variables. Permutation importance is the more trustworthy alternative when the ranking matters.

```python
feature_imp = pd.DataFrame(
    {'importance': RF_boston.feature_importances_},
    index=feature_names)
feature_imp.sort_values(by='importance', ascending=False)
```

## 8.3.4 Boosting

Boosting inverts the logic of bagging. No resampling: trees are grown **sequentially**, each
fitted to the residuals left by the ones before, and shrunk by a learning rate $\lambda$ before
being added:

$$\hat f(x) \leftarrow \hat f(x) + \lambda \hat f^b(x).$$

Learning slowly is the point. Three parameters interact:

- `n_estimators` ($B$) unlike bagging, boosting **can** overfit if $B$ is too large.
- `learning_rate` ($\lambda$) smaller needs larger $B$; the two trade off directly.
- `max_depth` ($d$) the interaction order. Depth 1 (stumps) gives a purely additive model.

```python
boost_boston = GBR(n_estimators=5000,
                   learning_rate=0.001,
                   max_depth=3,
                   random_state=0)
boost_boston.fit(X_train, y_train)
```

`staged_predict()` replays the ensemble one tree at a time, so training and test error can be
tracked along the boosting path. Training error decreases monotonically by construction; the
test curve is the one that matters. If it turns upward, the ensemble is overfitting and $B$
is too large that gap between the curves is the visual definition of the bias-variance
trade-off in boosting.

```python
test_error = np.zeros_like(boost_boston.train_score_)
for idx, y_ in enumerate(boost_boston.staged_predict(X_test)):
    test_error[idx] = np.mean((y_test - y_)**2)

plot_idx = np.arange(boost_boston.train_score_.shape[0])
ax = subplots(figsize=(8, 8))[1]
ax.plot(plot_idx,
        boost_boston.train_score_,
        'b',
        label='Training')
ax.plot(plot_idx,
        test_error,
        'r',
        label='Test')
ax.set_xlabel('Number of trees')
ax.set_ylabel('MSE')
ax.legend();
```

Test MSE around 14.48  comparable to bagging, and well ahead of the single pruned tree.

```python
y_hat_boost = boost_boston.predict(X_test)
np.mean((y_test - y_hat_boost)**2)
```

Raising $\lambda$ from 0.001 to 0.2 gives essentially the same test MSE (14.50). With 5000
trees available, both settings have enough capacity to converge the larger rate simply gets
there sooner. On a harder problem, or with fewer trees, the difference would be substantial, and
$\lambda$ and $B$ should be tuned jointly.

```python
boost_boston = GBR(n_estimators=5000,
                   learning_rate=0.2,
                   max_depth=3,
                   random_state=0)
boost_boston.fit(X_train,
                 y_train)
y_hat_boost = boost_boston.predict(X_test)
np.mean((y_test - y_hat_boost)**2)
```

## 8.3.5 Bayesian Additive Regression Trees

BART also builds trees successively, but each iteration *perturbs* an existing tree rather than
fitting a fresh one to residuals a Markov chain over sums of trees. Perturbation helps escape
local optima, and because the output is a posterior sample, uncertainty intervals come for free.

`burnin` discards the early draws before the chain has settled; `ndraw` is the number of
posterior samples kept. The values below are deliberately tiny so the cell runs quickly
a serious fit would use something like `burnin=100, ndraw=1000`. Expect this cell to be the
slowest in the notebook.

`BART()` here handles quantitative outcomes only; logistic and probit variants exist in other
implementations.

```python
bart_boston = BART(random_state=0, burnin=5, ndraw=15)
bart_boston.fit(X_train, y_train)
```

Test MSE around 20.92, similar to the random forest. Note the explicit `astype(np.float32)`
the BART implementation expects that dtype and will complain otherwise. With only 15 draws this
number is noisy; a longer chain would give a fairer comparison.

```python
yhat_test = bart_boston.predict(X_test.astype(np.float32))
np.mean((y_test - yhat_test)**2)
```

BART's analogue of variable importance is how often each variable was selected for a split,
averaged across posterior draws. The ordering is broadly consistent with the random forest
`rm` and `lstat` on top though the spread is much narrower, since BART keeps every variable
in play rather than concentrating on a few.

```python
var_inclusion = pd.Series(bart_boston.variable_inclusion_.mean(0),
                          index=D.columns)
var_inclusion
```

## Summary

**Test MSE on `Boston`** (same 70/30 split throughout):

| Model | Test MSE | RMSE |
|---|---|---|
| Pruned regression tree | 28.07 | 5.30 |
| Bagging (100 trees) | 14.63 | 3.82 |
| Bagging (500 trees) | 14.61 | 3.82 |
| Random forest (`max_features=6`) | 20.04 | 4.48 |
| Boosting ($\lambda = 0.001$) | 14.48 | 3.81 |
| Boosting ($\lambda = 0.2$) | 14.50 | 3.81 |
| BART (15 draws) | 20.92 | 4.57 |

Halving the error of a single tree is the real payoff of ensembling. These are single-split
estimates, though, so small gaps 14.48 against 14.63 are not meaningful differences;
repeated splits or nested CV would be needed to rank the ensembles with confidence.

**What generalizes beyond this dataset**

- A single tree is unmatched for explaining a model to a non-technical audience and mediocre
  at predicting. Ensembles reverse both.
- Bagging and random forests are safe in `n_estimators` but sensitive to `max_features`
  and, as seen here, the forest can lose to plain bagging when a few predictors dominate.
- Boosting is the opposite: it can overfit with too many trees, so $B$ and $\lambda$ must be
  tuned together and the staged test curve is the diagnostic to watch.
- Trees need no feature scaling and handle mixed predictor types natively, which makes them a
  strong default on tabular data.

**Practices carried through this lab**

- The test half was split off before any pruning or tuning, and cross-validation ran only on
  the training half the discipline that makes the reported numbers believable.
- Cross-validation selecting a *worse* model (the pruned classification tree) is reported
  rather than smoothed over. It is unbiased for selection but not free of variance.
- Impurity-based importance is a useful ranking with a known bias, not ground truth.
